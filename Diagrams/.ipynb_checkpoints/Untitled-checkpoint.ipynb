```python
from diagrams import Diagram, Cluster, Edge
from diagrams.aws.compute import EC2, EC2AutoScaling, ECS, EKS, Lambda
from diagrams.aws.database import DynamoDB, RDS, ElastiCache, Neptune
from diagrams.aws.storage import S3, Glacier
from diagrams.aws.network import ELB, ALB, NLB, CloudFront, VPC, PrivateSubnet, PublicSubnet, WAF
from diagrams.aws.iot import IotCore
from diagrams.aws.analytics import Kinesis, MSK
from diagrams.aws.integration import SQS, SNS
from diagrams.aws.ml import Sagemaker
from diagrams.aws.security import KMS, IAM
from diagrams.aws.management import Cloudwatch
from diagrams.onprem.compute import Server
from diagrams.onprem.iot import Iot
from diagrams.programming.language import Python
from diagrams.generic.compute import Rack
from diagrams.generic.storage import Storage
from diagrams.generic.network import Router

# Create the system architecture diagram
with Diagram("Smart Traffic Management System Architecture", show=False, direction="TB"):
    
    # Edge Devices Layer
    with Cluster("Edge Devices"):
        traffic_cameras = [Iot("Camera 1"), Iot("Camera 2"), Iot("Camera N")]
        iot_sensors = [Iot("Sensor 1"), Iot("Sensor 2")]
        edge_devices = traffic_cameras + iot_sensors
    
    # Network & Security Layer
    with Cluster("AWS Cloud - Network & Security"):
        with Cluster("VPC"):
            with Cluster("Public Subnet"):
                alb = ALB("Application\nLoad Balancer\n(HTTP/REST)")
                nlb = NLB("Network\nLoad Balancer\n(TCP/RTSP)")
            
            with Cluster("Private Subnets"):
                waf = WAF("AWS WAF")
                internet_gw = Router("Internet Gateway")
        
        security_group = IAM("IAM Least Privilege")
        kms = KMS("KMS Encryption")
    
    # Message Ingestion Layer
    with Cluster("Message Ingestion & Queuing"):
        iot_core = IotCore("AWS IoT Core\n(MQTT)")
        kafka = MSK("Kafka/MSK\nHigh Throughput")
        kinesis = Kinesis("Kinesis\nStream Processing")
        sqs = SQS("SQS\nSimple Queueing")
        sns = SNS("SNS\nNotifications")
    
    # Compute Layer
    with Cluster("Compute & Auto Scaling"):
        with Cluster("GPU Inference Tier"):
            gpu_asg = EC2AutoScaling("GPU Auto Scaling\n(g4/g5 instances)")
            gpu_instances = [EC2("GPU Instance 1"), EC2("GPU Instance 2")]
            gpu_asg >> gpu_instances
        
        with Cluster("CPU Processing Tier"):
            cpu_asg = EC2AutoScaling("CPU Auto Scaling")
            cpu_instances = [EC2("CPU Instance 1"), EC2("CPU Instance 2")]
            cpu_asg >> cpu_instances
        
        lambda_func = Lambda("Serverless\nFunctions")
    
    # Data Storage Layer
    with Cluster("Data Storage & Databases"):
        with Cluster("Primary Databases"):
            cassandra = DynamoDB("Cassandra/DynamoDB\nTime-series & High Write")
            mongodb = DynamoDB("MongoDB Atlas\nFlexible Queries & Geospatial")
        
        with Cluster("Caching Layer"):
            redis = ElastiCache("Redis/ElastiCache\nFast-access Metadata")
        
        with Cluster("Object Storage"):
            s3 = S3("S3 Bucket\nVideo & Images")
            glacier = Glacier("Glacier\nCold Storage")
            s3_lifecycle = Edge(label="Lifecycle Rules") >> glacier
    
    # MLOps & Model Management
    with Cluster("MLOps & Model Operations"):
        mlflow = Sagemaker("Model Registry\n(MLflow/SageMaker)")
        cicd_pipeline = ECS("CI/CD Pipeline\nBlue-Green Deployments")
        model_monitoring = Cloudwatch("Model Performance\nMonitoring")
    
    # Monitoring & Observability
    with Cluster("Monitoring & Observability"):
        cloudwatch = Cloudwatch("CloudWatch\nMetrics & Logs")
        metrics = [
            "CPU/GPU Utilization",
            "Latency Metrics",
            "Queue Depth",
            "Instance Count",
            "Auto Scaling Events"
        ]
    
    # Data Flow Connections
    
    # Edge to Cloud Connectivity
    edge_devices >> iot_core
    edge_devices >> nlb
    
    # Load Balancer to Compute
    alb >> cpu_asg
    nlb >> gpu_asg
    
    # Message Queue to Processing
    iot_core >> kinesis
    kinesis >> cpu_asg
    kafka >> gpu_asg
    sqs >> lambda_func
    
    # Compute to Storage
    cpu_instances >> cassandra
    cpu_instances >> mongodb
    gpu_instances >> redis
    
    # Storage relationships
    cpu_instances >> s3
    gpu_instances >> s3
    
    # MLOps connections
    mlflow >> cicd_pipeline
    cicd_pipeline >> gpu_asg
    cicd_pipeline >> cpu_asg
    
    # Monitoring connections
    cloudwatch >> alb
    cloudwatch >> nlb
    cloudwatch >> gpu_asg
    cloudwatch >> cpu_asg
    cloudwatch >> kinesis
    cloudwatch >> kafka
    
    # Security connections
    waf >> alb
    waf >> nlb
    kms >> s3
    kms >> cassandra
    kms >> mongodb

# Additional detailed diagram for scaling strategy
with Diagram("Scaling Strategy Details", show=False, direction="TB"):
    
    with Cluster("Horizontal Scaling (Preferred)"):
        with Cluster("Auto Scaling Groups"):
            asg_cpu = EC2AutoScaling("CPU ASG")
            asg_gpu = EC2AutoScaling("GPU ASG")
        
        scaling_metrics = [
            "CPU Utilization > 70%",
            "GPU Memory > 80%",
            "Queue Depth > 1000",
            "Latency > 200ms"
        ]
        
        scaling_actions = [
            "Add 2 instances",
            "Remove 1 instance",
            "Scale to 10 max"
        ]
    
    with Cluster("Vertical Scaling (Manual)"):
        manual_scaling = EC2("Current Instance")
        upgrade_process = [
            "Stop Instance",
            "Change Type",
            "Restart Instance"
        ]
        new_instance = EC2("Larger Instance")
    
    with Cluster("Cost Optimization"):
        spot_instances = EC2("Spot Instances\n(60-90% savings)")
        reserved_instances = EC2("Reserved Instances")
        savings_plan = Lambda("Savings Plan")
    
    # Connections
    scaling_metrics >> asg_cpu
    scaling_metrics >> asg_gpu
    asg_cpu >> scaling_actions
    asg_gpu >> scaling_actions
    
    manual_scaling >> upgrade_process >> new_instance

# Database Selection Diagram
with Diagram("Database Strategy", show=False, direction="LR"):
    
    with Cluster("Time-series & High Write"):
        cassandra_db = DynamoDB("Cassandra")
        dynamodb = DynamoDB("DynamoDB")
        features_ts = [
            "High write throughput",
            "Time-series optimized",
            "Auto-scaling"
        ]
    
    with Cluster("Flexible Query & Geospatial"):
        mongodb = DynamoDB("MongoDB Atlas")
        features_geo = [
            "Rich queries",
            "Geospatial indexing",
            "Document model"
        ]
    
    with Cluster("Caching Layer"):
        redis_cache = ElastiCache("Redis")
        cache_features = [
            "Sub-millisecond latency",
            "Congestion counters",
            "Metadata cache"
        ]
    
    # Usage patterns
    write_heavy = Storage("Traffic Events\n1000+ writes/sec")
    query_flexible = Storage("Geospatial Queries\nComplex filtering")
    fast_access = Storage("Real-time Data\n<1ms response")
    
    write_heavy >> cassandra_db
    write_heavy >> dynamodb
    query_flexible >> mongodb
    fast_access >> redis_cache

print("System architecture diagrams generated successfully!")
print("\nKey Architecture Components:")
print("1. Edge Devices → IoT Core/NLB for connectivity")
print("2. Multiple Message Queues (Kafka/Kinesis/SQS) for different workloads")
print("3. Separate GPU/CPU Auto Scaling Groups for cost optimization")
print("4. Multi-database strategy (Cassandra + MongoDB + Redis)")
print("5. Comprehensive monitoring with CloudWatch")
print("6. MLOps with model registry and CI/CD")
print("7. Security with WAF, KMS, and IAM")

print("\nScaling Strategy:")
print("- Horizontal: Auto Scaling Groups with metrics-based triggers")
print("- Vertical: Manual instance type upgrades")
print("- Cost: Spot instances, reserved instances, savings plans")

print("\nDatabase Selection:")
print("- Cassandra/DynamoDB: Time-series, high write throughput")
print("- MongoDB Atlas: Flexible queries, geospatial data")
print("- Redis: Caching, fast-access metadata")
```
