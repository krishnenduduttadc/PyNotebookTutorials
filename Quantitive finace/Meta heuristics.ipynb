Simulated Annealing (simple & classic)

```python
import random
import math

def objective(x):
    return x**2  # minimize

def simulated_annealing():
    x = random.uniform(-10, 10)
    T = 1000
    cooling = 0.99

    while T > 1e-3:
        x_new = x + random.uniform(-1, 1)
        delta = objective(x_new) - objective(x)

        if delta < 0 or random.random() < math.exp(-delta / T):
            x = x_new

        T *= cooling

    return x

print(simulated_annealing())
```

Genetic Algorithm (most famous)

```python
import random

POP_SIZE = 20
GENES = 1

def fitness(x):
    return -x**2  # maximize

population = [random.uniform(-10, 10) for _ in range(POP_SIZE)]

for _ in range(100):
    population.sort(key=fitness, reverse=True)
    next_gen = population[:5]  # elitism

    while len(next_gen) < POP_SIZE:
        parent = random.choice(population[:10])
        child = parent + random.uniform(-1, 1)
        next_gen.append(child)

    population = next_gen

best = max(population, key=fitness)
print(best)
```

Particle Swarm Optimization (PSO)

```python
import random

particles = [random.uniform(-10, 10) for _ in range(10)]
velocities = [0]*10
best_global = min(particles, key=lambda x: x**2)

for _ in range(100):
    for i in range(len(particles)):
        velocities[i] = 0.5 * velocities[i] + random.random() * (best_global - particles[i])
        particles[i] += velocities[i]

    best_global = min(particles, key=lambda x: x**2)

print(best_global)
```
