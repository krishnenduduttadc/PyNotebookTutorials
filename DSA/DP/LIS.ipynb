# Longest Increasing Subsequence (LIS)

Given an array $arr[0..n-1]$, the **Longest Increasing Subsequence (LIS)** is the longest subsequence where all elements are strictly increasing.

---

## Dynamic Programming Formulation

Let $dp[i]$ = length of LIS ending at index $i$.

### Recurrence:
\[
dp[i] = 1 + \max \{ dp[j] \;|\; j < i \text{ and } arr[i] > arr[j] \}
\]

If no such $j$ exists, then:
\[
dp[i] = 1
\]

### Final Answer:
\[
\text{LIS}(arr) = \max_{0 \leq i < n} dp[i]
\]


```python
def lis_visual(arr):
    n = len(arr)
    dp = [0] * n
    omax = 0  # overall max

    def print_dp(step):
        print(f"Step {step}:")
        print("Index:", list(range(n)))
        print("Array:", arr)
        print(" DP  :", dp)
        print("Overall Max LIS so far:", omax)
        print("-" * 50)

    for i in range(n):
        max_val = 0
        for j in range(i):
            if arr[i] > arr[j]:
                max_val = max(max_val, dp[j])
        dp[i] = max_val + 1
        omax = max(omax, dp[i])

        # print after each update
        print_dp(i)

    print("Length of LIS:", omax)
    print("Final DP array:", dp)
    return omax


# Example usage
lis_visual([10, 22, 9, 33, 21, 50, 41, 60, 80, 3])
```
