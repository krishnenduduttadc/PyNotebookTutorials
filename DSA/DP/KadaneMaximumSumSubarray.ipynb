```python
def kadane_maximum_sum_subarray(arr):
    csum = arr[0]  # current sum
    osum = arr[0]  # overall max sum

    print(f"Start: csum = {csum}, osum = {osum}\n")

    for i in range(1, len(arr)):
        print(f"Step {i}: processing arr[{i}] = {arr[i]}")

        if csum >= 0:
            csum += arr[i]
            reason = "extend current subarray"
        else:
            csum = arr[i]
            reason = "restart from current element"

        if csum > osum:
            osum = csum
            updated = "update overall max"
        else:
            updated = "no update to overall max"

        print(f"  -> csum = {csum} ({reason}), osum = {osum} ({updated})\n")

    print(f"Final Maximum Subarray Sum = {osum}")
    return osum


# Example run
kadane_maximum_sum_subarray([5, 6, 7, 4, 3, 6, 4])
```
