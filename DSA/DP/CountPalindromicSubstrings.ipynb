```python
def count_palindromic_substrings(s: str) -> int:
    n = len(s)
    dp = [[False] * n for _ in range(n)]
    count = 0

    print("--- Filling DP table ---\n")

    for g in range(n):  # gap = substring length - 1
        for i in range(n - g):
            j = i + g
            if g == 0:
                dp[i][j] = True
                reason = f"single char '{s[i]}'"
            elif g == 1:
                if s[i] == s[j]:
                    dp[i][j] = True
                    reason = f"two same chars '{s[i]}{s[j]}'"
                else:
                    dp[i][j] = False
                    reason = f"two diff chars '{s[i]}{s[j]}'"
            else:
                if s[i] == s[j] and dp[i + 1][j - 1]:
                    dp[i][j] = True
                    reason = f"ends match '{s[i]}..{s[j]}' and middle is palindrome"
                else:
                    dp[i][j] = False
                    reason = f"ends differ or middle not palindrome"

            if dp[i][j]:
                count += 1
                print(f"Palindrome found: \"{s[i:j+1]}\" (dp[{i}][{j}] = True, {reason})")
            else:
                print(f"Not palindrome: \"{s[i:j+1]}\" (dp[{i}][{j}] = False, {reason})")

        # Show DP table after finishing this gap
        print(f"\nDP after gap={g}:")
        for row in dp:
            print(row)
        print("-" * 40)

    print("\nFinal DP table:")
    for row in dp:
        print(row)

    print(f"\nTotal palindromic substrings: {count}")
    return count


# Example run
count_palindromic_substrings("abccbc")
```
