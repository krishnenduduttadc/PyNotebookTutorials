```python
def rod_cutting(prices):
    n = len(prices)
    
    # shift prices by 1 (so np[i] = price for rod of length i)
    np = [0] * (n + 1)
    for i in range(n):
        np[i + 1] = prices[i]

    dp = [0] * (n + 1)
    dp[0] = 0
    dp[1] = np[1]

    print("--- Filling DP table ---\n")
    print(f"dp[1] = {dp[1]} (price of rod length 1)")

    for i in range(2, n + 1):
        dp[i] = np[i]  # start with selling the rod as a whole
        print(f"\nEvaluating rod length {i}: base price = {np[i]}")

        li, ri = 1, i - 1
        while li <= ri:
            combined = dp[li] + dp[ri]
            if combined > dp[i]:
                dp[i] = combined
                print(f"  Cut into {li} + {ri} → {dp[li]} + {dp[ri]} = {combined} (new best)")
            else:
                print(f"  Cut into {li} + {ri} → {dp[li]} + {dp[ri]} = {combined} (not better)")

            li += 1
            ri -= 1

        print(f"Best price for rod length {i} = {dp[i]}")
        print("DP state:", dp)

    print("\nFinal DP table:", dp)
    print(f"\nMaximum Obtainable Price = {dp[n]}")
    return dp[n]


# Example run
rod_cutting([1, 5, 8, 9, 10, 17, 17, 20])
```
