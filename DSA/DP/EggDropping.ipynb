# Egg Dropping Problem

We are given:
- $n$ eggs
- $k$ floors

Goal: Find the minimum number of trials needed in the worst case to determine the highest floor from which an egg can be dropped without breaking.

---

## Recurrence Relation

Let $dp[i][j]$ = minimum number of attempts needed with $i$ eggs and $j$ floors.

### Base Cases:
\[
dp[1][j] = j \quad \text{(only one egg → must try all floors)}
\]
\[
dp[i][1] = 1 \quad \text{(only one floor → one attempt needed)}
\]

### Transition:
If we drop an egg from floor $x$:
- If the egg **breaks**, we must solve for $(i-1, x-1)$ floors below.
- If the egg **survives**, we must solve for $(i, j-x)$ floors above.

So, the recurrence is:
\[
dp[i][j] = 1 + \min_{1 \leq x \leq j} \max \big( dp[i-1][x-1], dp[i][j-x] \big)
\]

---

### Final Answer:
\[
\text{EggDrop}(n, k) = dp[n][k]
\]


```python
def egg_drop_visual(n, k):
    # dp[i][j] = min trials with i eggs and j floors
    dp = [[0] * (k + 1) for _ in range(n + 1)]

    # Fill the table
    for i in range(1, n + 1):  # eggs
        for j in range(1, k + 1):  # floors
            if i == 1:
                dp[i][j] = j  # must try all floors
            elif j == 1:
                dp[i][j] = 1  # only 1 floor
            else:
                min_val = float("inf")
                # try dropping from each floor
                for mj in range(j - 1, -1, -1):
                    pj = j - 1 - mj
                    v1 = dp[i][mj]     # egg survives
                    v2 = dp[i-1][pj]   # egg breaks
                    val = max(v1, v2)
                    min_val = min(min_val, val)
                dp[i][j] = min_val + 1

            # Print intermediate DP state
            print(f"After computing dp[{i}][{j}] (eggs={i}, floors={j}):")
            for row in dp:
                print(row)
            print("-" * 40)

    print("Minimum trials needed:", dp[n][k])
    return dp[n][k]


# Example usage
egg_drop_visual(2, 4)
```
