```python
def wildcard_match(s: str, p: str) -> bool:
    # dp[i][j] means: does p[i:] match s[j:]
    dp = [[False] * (len(s) + 1) for _ in range(len(p) + 1)]

    # Fill table bottom-up
    for i in range(len(p), -1, -1):
        for j in range(len(s), -1, -1):
            if i == len(p) and j == len(s):
                dp[i][j] = True
            elif i == len(p):
                dp[i][j] = False
            elif j == len(s):
                dp[i][j] = (p[i] == '*') and dp[i + 1][j]
            else:
                if p[i] == '?':
                    dp[i][j] = dp[i + 1][j + 1]
                elif p[i] == '*':
                    dp[i][j] = dp[i + 1][j] or dp[i][j + 1]
                else:
                    dp[i][j] = (p[i] == s[j]) and dp[i + 1][j + 1]

        # 🔍 Visualization after each row of pattern processed
        print(f"\nAfter processing pattern[{i}] = '{p[i] if i < len(p) else 'END'}':")
        for row in dp:
            print(" ".join(['T' if x else 'F' for x in row]))

    return dp[0][0]


# Example run
s1 = "baaabab"
s2 = "ba*a"
print("\nFinal Result:", wildcard_match(s1, s2))
```
