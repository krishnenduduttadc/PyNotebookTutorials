```python
def longest_bitonic_subsequence(arr):
    n = len(arr)

    # Step 1: Compute LIS (Longest Increasing Subsequence) from left
    lis = [0] * n
    for i in range(n):
        max_val = 0
        for j in range(i):
            if arr[j] <= arr[i]:
                max_val = max(max_val, lis[j])
        lis[i] = max_val + 1
        print(f"LIS[{i}] (arr[{i}]={arr[i]}) = {lis[i]}, LIS so far: {lis}")

    print("\nLIS array:", lis)

    # Step 2: Compute LDS (Longest Decreasing Subsequence) from right
    lds = [0] * n
    for i in range(n - 1, -1, -1):
        max_val = 0
        for j in range(n - 1, i, -1):
            if arr[j] <= arr[i]:
                max_val = max(max_val, lds[j])
        lds[i] = max_val + 1
        print(f"LDS[{i}] (arr[{i}]={arr[i]}) = {lds[i]}, LDS so far: {lds}")

    print("\nLDS array:", lds)

    # Step 3: Find maximum LIS[i] + LDS[i] - 1
    omax = 0
    for i in range(n):
        bitonic_len = lis[i] + lds[i] - 1
        omax = max(omax, bitonic_len)
        print(f"At index {i}, arr[{i}]={arr[i]}, LIS={lis[i]}, LDS={lds[i]} → Bitonic={bitonic_len}")

    print(f"\nLongest Bitonic Subsequence length = {omax}")
    return omax


# Example run
longest_bitonic_subsequence([10, 22, 9, 33, 21, 50, 41, 60, 80, 3])
```
