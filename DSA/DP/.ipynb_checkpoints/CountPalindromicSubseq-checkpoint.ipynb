```python
def count_palindromic_subsequences(s: str) -> int:
    n = len(s)
    dp = [[0] * n for _ in range(n)]

    print("--- Filling DP table ---\n")

    for g in range(n):  # gap = substring length - 1
        for i in range(n - g):
            j = i + g
            if g == 0:
                dp[i][j] = 1
                reason = "single char → palindrome itself"
            elif g == 1:
                if s[i] == s[j]:
                    dp[i][j] = 2
                    reason = f"two same chars '{s[i]}{s[j]}' → 2 palindromes"
                else:
                    dp[i][j] = 1
                    reason = f"two diff chars '{s[i]}{s[j]}' → only 1 palindrome"
            else:
                if s[i] == s[j]:
                    dp[i][j] = dp[i][j - 1] + dp[i + 1][j] + 1
                    reason = f"ends match '{s[i]}..{s[j]}' → add +1"
                else:
                    dp[i][j] = dp[i][j - 1] + dp[i + 1][j] - dp[i + 1][j - 1]
                    reason = f"ends differ '{s[i]}..{s[j]}' → inclusion-exclusion"

            print(f"dp[{i}][{j}] (\"{s[i:j+1]}\") = {dp[i][j]} ({reason})")

        # Show DP after this gap is completed
        print(f"\nDP after gap={g}:")
        for row in dp:
            print(row)
        print("-" * 40)

    print("\nFinal DP table:")
    for row in dp:
        print(row)

    print(f"\nTotal palindromic subsequences: {dp[0][n-1]}")
    return dp[0][n - 1]


# Example run
count_palindromic_subsequences("abccbc")
```
