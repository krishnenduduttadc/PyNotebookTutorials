```python
def count_distinct_subsequences(s: str):
    n = len(s)
    dp = [0] * (n + 1)
    dp[0] = 1  # empty subsequence
    last_occurrence = {}

    print("Initial DP state:", dp)
    print("Last occurrence map:", last_occurrence)

    for i in range(1, n + 1):
        dp[i] = 2 * dp[i - 1]
        ch = s[i - 1]
        print(f"\nStep {i}: processing '{ch}' (index {i - 1} in string)")
        print(f"  Base: dp[{i}] = 2 * dp[{i-1}] = 2 * {dp[i-1]} = {dp[i]}")

        if ch in last_occurrence:
            j = last_occurrence[ch]
            deduction = dp[j - 1]
            dp[i] -= deduction
            print(f"  '{ch}' seen before at position {j}")
            print(f"  Deduct dp[{j-1}] = {deduction}, so dp[{i}] = {dp[i]}")
        else:
            print(f"  '{ch}' is new, no deduction.")

        last_occurrence[ch] = i
        print(f"  Updated last occurrence: {last_occurrence}")
        print(f"  DP state now: {dp}")

    result = dp[n] - 1  # exclude empty subsequence
    print("\nFinal DP table:", dp)
    print(f"Number of distinct subsequences (excluding empty): {result}")
    return result


# Example run
count_distinct_subsequences("abc")
```
