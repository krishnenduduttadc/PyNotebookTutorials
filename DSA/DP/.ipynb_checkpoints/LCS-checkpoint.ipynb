# Longest Common Subsequence (LCS)

Given two strings $s_1$ and $s_2$, the **Longest Common Subsequence (LCS)** is the longest sequence of characters that appears in both strings **in the same order**, but not necessarily contiguously.

---

## Dynamic Programming Formulation

Let $dp[i][j]$ represent the length of the LCS of the substrings:
\[
s_1[i..] \quad \text{and} \quad s_2[j..]
\]

### Base Case:
If either string is exhausted:
\[
dp[i][j] = 0 \quad \text{when } i = |s_1| \text{ or } j = |s_2|
\]

### Transition:
If the current characters match:
\[
dp[i][j] = 1 + dp[i+1][j+1]
\]

Otherwise:
\[
dp[i][j] = \max(dp[i+1][j], \; dp[i][j+1])
\]

---

### Final Answer:
\[
\text{LCS}(s_1, s_2) = dp[0][0]
\]


```python
def lcs_visual(s1, s2):
    m, n = len(s1), len(s2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    def print_table():
        # Print header row with indices
        header = "    " + "  ".join([f"{j:2}" for j in range(n + 1)])
        print(header)
        for i in range(m + 1):
            row_label = f"{i:2} " if i < m else f"{i:2} "
            row_values = "  ".join([f"{dp[i][j]:2}" for j in range(n + 1)])
            print(row_label + " " + row_values)
        print("-" * (6 + 4 * (n + 1)))

    # Fill from bottom-right (like the Java code)
    for i in range(m - 1, -1, -1):
        for j in range(n - 1, -1, -1):
            if s1[i] == s2[j]:
                dp[i][j] = 1 + dp[i + 1][j + 1]
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j + 1])

            print(f"After processing s1[{i}]='{s1[i]}' and s2[{j}]='{s2[j]}':")
            print_table()

    print("Length of LCS:", dp[0][0])
    return dp[0][0]


# Example usage
lcs_visual("abcd", "abbd")
```
