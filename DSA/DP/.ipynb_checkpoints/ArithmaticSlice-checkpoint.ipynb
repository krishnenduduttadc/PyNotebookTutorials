```python
import matplotlib.pyplot as plt

def arithmetic_slices(arr):
    n = len(arr)
    dp = [0] * n
    ans = 0
    steps = []  # store snapshots for visualization

    for i in range(2, n):
        if arr[i] - arr[i-1] == arr[i-1] - arr[i-2]:
            dp[i] = dp[i-1] + 1
            ans += dp[i]
        steps.append((i, dp.copy(), ans))

    return ans, steps


def visualize_steps(arr, steps):
    print("Array:", arr)
    print("="*60)
    for idx, dp_state, ans in steps:
        print(f"Index {idx}: dp = {dp_state}, Total Arithmetic Slices = {ans}")

        # plot dp state at this step
        plt.figure(figsize=(8, 3))
        plt.bar(range(len(dp_state)), dp_state, color="skyblue")
        plt.title(f"Step {idx}: dp[{idx}] updated, total slices = {ans}")
        plt.xlabel("Index")
        plt.ylabel("dp[i] (number of slices ending at i)")
        plt.show()


# Example usage
arr = [2, 5, 9, 12, 15, 18, 22, 26, 30, 34, 36, 38, 40, 41]
ans, steps = arithmetic_slices(arr)
print("Final Answer:", ans)
visualize_steps(arr, steps)
```
