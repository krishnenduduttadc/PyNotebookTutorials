### Burst Balloons (Interval DP)

We are given an array of balloons with values 
$nums = [a_1, a_2, \dots, a_n]$.

When we burst balloon $k$, we gain coins:
\[
coins = nums[i-1] \times nums[k] \times nums[j+1]
\]
where $i-1$ and $j+1$ are neighbors of $k$ in the remaining array.

---

#### DP Definition

Let
\[
dp[i][j] = \text{maximum coins from bursting all balloons in interval } [i, j]
\]

#### Recurrence
\[
dp[i][j] = \max_{k=i}^{j} \Big( dp[i][k-1] + dp[k+1][j] + nums[i-1] \cdot nums[k] \cdot nums[j+1] \Big)
\]

Boundary conditions:
- If $i > j$, then $dp[i][j] = 0$.
- We add **virtual balloons** $nums[0] = nums[n+1] = 1$.

#### Final Answer
\[
dp[1][n] \quad \text{(bursting all balloons from 1 to n)}
\]


```python
def burst_balloons(nums):
    n = len(nums)
    arr = [1] + nums + [1]  # add virtual balloons
    dp = [[0] * (n+2) for _ in range(n+2)]
    steps = []

    for g in range(1, n+1):  # gap
        for i in range(1, n-g+2):
            j = i + g - 1
            max_val = float("-inf")
            chosen_k = None
            for k in range(i, j+1):
                left = dp[i][k-1]
                right = dp[k+1][j]
                val = arr[i-1] * arr[k] * arr[j+1]
                total = left + right + val
                if total > max_val:
                    max_val = total
                    chosen_k = k
            dp[i][j] = max_val
            steps.append((i, j, chosen_k, max_val, [row[:] for row in dp]))

    return dp[1][n], steps


def visualize_textual(steps):
    for (i, j, k, val, dp_state) in steps:
        print("="*50)
        print(f"Interval [{i}, {j}] → Best k = {k}, Coins = {val}")
        print("DP Table:")
        for r in range(1, len(dp_state)-1):
            row_vals = [f"{dp_state[r][c]:3d}" for c in range(1, len(dp_state)-1)]
            print(" ".join(row_vals))
        print()


# Example
nums = [2, 3, 5]
ans, steps = burst_balloons(nums)
print("Final Answer:", ans)
visualize_textual(steps)
```
