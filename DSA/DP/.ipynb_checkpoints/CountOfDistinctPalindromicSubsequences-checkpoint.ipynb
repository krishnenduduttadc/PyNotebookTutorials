```python
def count_of_distinct_palindromic_subsequences(s: str) -> int:
    MOD = 10**9 + 7
    n = len(s)
    dp = [[0] * n for _ in range(n)]

    print("--- Filling DP table ---\n")

    for g in range(n):  # gap (length of substring - 1)
        for i in range(n - g):
            j = i + g
            if g == 0:
                dp[i][j] = 1
            else:
                if s[i] == s[j]:
                    l, r = i + 1, j - 1
                    while l <= r and s[l] != s[i]:
                        l += 1
                    while l <= r and s[r] != s[i]:
                        r -= 1

                    if l > r:  # no duplicate inside
                        dp[i][j] = 2 * dp[i + 1][j - 1] + 2
                        reason = "no duplicate inside"
                    elif l == r:  # one duplicate inside
                        dp[i][j] = 2 * dp[i + 1][j - 1] + 1
                        reason = "one duplicate inside"
                    else:  # two or more duplicates inside
                        dp[i][j] = 2 * dp[i + 1][j - 1] - dp[l + 1][r - 1]
                        reason = "two+ duplicates inside"
                else:
                    dp[i][j] = dp[i + 1][j] + dp[i][j - 1] - dp[i + 1][j - 1]
                    reason = "ends differ"

                dp[i][j] %= MOD

                print(f"dp[{i}][{j}] (\"{s[i:j+1]}\") = {dp[i][j]} ({reason})")

        # Show DP state after finishing this gap
        print(f"\nDP after gap={g}:")
        for row in dp:
            print(row)
        print("-" * 40)

    print("\nFinal DP table:")
    for row in dp:
        print(row)

    print(f"\nCount of distinct palindromic subsequences: {dp[0][n-1]}")
    return dp[0][n - 1]


# Example run
count_of_distinct_palindromic_subsequences("abccba")
```
