```python
def count_distinct_palindromic_subsequences(s: str) -> int:
    n = len(s)
    dp = [[0] * n for _ in range(n)]

    # Build prev array
    prev = [-1] * n
    last_seen = {}
    for i, ch in enumerate(s):
        prev[i] = last_seen.get(ch, -1)
        last_seen[ch] = i

    # Build next array
    next_ = [-1] * n
    last_seen.clear()
    for i in range(n - 1, -1, -1):
        ch = s[i]
        next_[i] = last_seen.get(ch, -1)
        last_seen[ch] = i

    print("Prev array:", prev)
    print("Next array:", next_)
    print("\n--- Filling DP table ---")

    # Fill DP table by gap
    for g in range(n):
        for i in range(n - g):
            j = i + g
            if g == 0:
                dp[i][j] = 1
            elif g == 1:
                dp[i][j] = 2
            else:
                sc, ec = s[i], s[j]
                if sc != ec:
                    dp[i][j] = dp[i + 1][j] + dp[i][j - 1] - dp[i + 1][j - 1]
                else:
                    n_idx = next_[i]
                    p_idx = prev[j]
                    if n_idx > p_idx:  # no same char inside
                        dp[i][j] = 2 * dp[i + 1][j - 1] + 2
                    elif n_idx == p_idx:  # one same char inside
                        dp[i][j] = 2 * dp[i + 1][j - 1] + 1
                    else:  # more than one same char inside
                        dp[i][j] = 2 * dp[i + 1][j - 1] - dp[n_idx + 1][p_idx - 1]

            print(f"dp[{i}][{j}] ({s[i:j+1]}): {dp[i][j]}")

        # Pretty print DP state for this gap
        print(f"\nDP after gap={g}:")
        for row in dp:
            print(row)
        print("-" * 40)

    result = dp[0][n - 1]
    print("\nFinal DP table:")
    for row in dp:
        print(row)
    print(f"\nTotal distinct palindromic subsequences: {result}")
    return result


# Example run
count_distinct_palindromic_subsequences("ababda")
```
