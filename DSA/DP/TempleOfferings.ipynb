```python
def temple_offerings(height):
    n = len(height)
    
    # Left-to-right pass
    larr = [0] * n
    larr[0] = 1
    for i in range(1, n):
        if height[i] > height[i - 1]:
            larr[i] = larr[i - 1] + 1
        else:
            larr[i] = 1
    print("Left-to-right offerings:", larr)

    # Right-to-left pass
    rarr = [0] * n
    rarr[-1] = 1
    for i in range(n - 2, -1, -1):
        if height[i] > height[i + 1]:
            rarr[i] = rarr[i + 1] + 1
        else:
            rarr[i] = 1
    print("Right-to-left offerings:", rarr)

    # Combine both passes
    ans = 0
    print("\nFinal offerings per temple:")
    for i in range(n):
        chosen = max(larr[i], rarr[i])
        ans += chosen
        print(f"Temple {i} (height={height[i]}): max({larr[i]}, {rarr[i]}) = {chosen}")

    print(f"\nTotal Offerings = {ans}")
    return ans


# Example run
temple_offerings([2, 3, 5, 6, 4, 8, 9])
```
