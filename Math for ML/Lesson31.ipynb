```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA

# 1. Generate some dummy data (100 samples, 5 features)
X, y = make_blobs(n_samples=100, n_features=5, random_state=42)

# 2. Initialize PCA to reduce data to 2 components
pca = PCA(n_components=2)

# 3. Fit the model and transform the data
X_reduced = pca.fit_transform(X)

# 4. Print the shape of the original and reduced data
print("Original shape:", X.shape)
print("Reduced shape: ", X_reduced.shape)
```
