```python
import numpy as np

N = 100000

x = np.random.uniform(-1,1,N)
y = np.random.uniform(-1,1,N)

inside = (x*x + y*y <= 1)

pi_estimate = 4*np.mean(inside)

print(pi_estimate)
```

```python
import numpy as np

samples = []

x = 0

for _ in range(10000):

    proposal = x + np.random.normal()

    ratio = np.exp(
        -proposal**2/2
    ) / np.exp(
        -x**2/2
    )

    if np.random.rand() < min(1, ratio):
        x = proposal

    samples.append(x)

print(np.mean(samples))
```
