```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([20,40,60,80,100])

w = 0
lr = 0.01

for _ in range(1000):

    y_pred = w*x

    grad = -2*np.mean(
        x*(y-y_pred)
    )

    w -= lr*grad

print(w)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([
    [1],
    [2],
    [3],
    [4]
])

y = np.array([
    20,
    40,
    60,
    80
])

model = LinearRegression()

model.fit(X,y)

print(model.coef_)
print(model.intercept_)
```
