```python
w = 10

lr = 0.1

for i in range(20):

    loss = (w - 5)**2

    gradient = 2 * (w - 5)

    w = w - lr * gradient

    print(
        f"Step={i:2d} "
        f"Weight={w:.4f} "
        f"Loss={loss:.4f}"
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt

w = np.linspace(-2, 12, 500)

loss = (w - 5)**2

plt.plot(w, loss)

plt.xlabel("Weight")
plt.ylabel("Loss")

plt.title("Loss Function")

plt.grid()

plt.show()
```
