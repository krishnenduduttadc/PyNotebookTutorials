```python
import numpy as np

p = np.array([0.5,0.5])

entropy = -np.sum(
    p*np.log2(p)
)

print(entropy)
```

```python
import numpy as np

p = np.array([1,0])

q = np.array([0.9,0.1])

cross_entropy = -np.sum(
    p*np.log(q)
)

print(cross_entropy)
```

```python
import numpy as np

p = np.array([0.5,0.5])

q = np.array([0.4,0.6])

kl = np.sum(
    p*np.log(p/q)
)

print(kl)
```
