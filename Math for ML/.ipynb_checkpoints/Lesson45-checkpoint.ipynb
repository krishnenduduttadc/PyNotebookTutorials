```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# 1. Generate a linearly separable dataset (2 classes, 2 features)
X, y = make_blobs(n_samples=150, centers=2, random_state=42, cluster_std=1.5)

# 2. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- Your original code ---
# 3. Initialize the Support Vector Classifier with a linear kernel
model = SVC(kernel='linear')

# 4. Fit the model to the training data
model.fit(X_train, y_train)

# 5. Predict on the test data
pred = model.predict(X_test)
# --------------------------

# 6. Evaluate the results
print("--- Linear SVM Performance ---")
print(f"Accuracy Score: {accuracy_score(y_test, pred):.2f}\n")
print("Classification Report:")
print(classification_report(y_test, pred))
```

```python
import numpy as np
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Generate a non-linear dataset (concentric circles)
X, y = make_circles(n_samples=200, factor=0.5, noise=0.1, random_state=42)

# 2. Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- Your original code ---
# 3. Initialize the model with RBF kernel
model = SVC(
    kernel='rbf',
    C=1.0,
    gamma='scale'
)

# 4. Fit the model
model.fit(X_train, y_train)
# --------------------------

# 5. Predict and Evaluate
predictions = model.predict(X_test)
print("--- RBF SVM Performance ---")
print(f"Test Accuracy: {accuracy_score(y_test, predictions):.2f}")
```
