```python
import torch

x = torch.tensor(
    3.0,
    requires_grad=True
)

y = x**2

y.backward()

print(x.grad)
```

```python
import torch

x = torch.tensor(
    2.0,
    requires_grad=True
)

y = torch.tensor(
    3.0,
    requires_grad=True
)

z = x + y
L = z**2

L.backward()

print(x.grad)
print(y.grad)
```
