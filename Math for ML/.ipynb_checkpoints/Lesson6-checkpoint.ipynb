```python
import numpy as np

data = np.random.normal(
    loc=100,
    scale=10,
    size=10000
)

print(np.mean(data))
print(np.std(data))
```

```python
import numpy as np

data = [10,20,30,40,50]

print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Std:", np.std(data))
```

```python
import numpy as np

height = [160,170,180]
weight = [50,60,70]

print(np.corrcoef(height, weight))
```
