```python
import numpy as np

A = np.array([
    [3,1],
    [1,3]
])

U, S, VT = np.linalg.svd(A)

print("U =")
print(U)

print("\nSingular Values =")
print(S)

print("\nVT =")
print(VT)
```

```python
import numpy as np

x = np.array([1,2,3,4,5])

mean = np.mean(x)
variance = np.var(x)
std = np.std(x)

print("Mean =", mean)
print("Variance =", variance)
print("Std =", std)
```
