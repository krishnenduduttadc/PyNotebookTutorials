```python
import numpy as np

samples = np.random.binomial(
    n=1,
    p=0.8,
    size=20
)

print(samples)
```

```python
import numpy as np

samples = np.random.binomial(
    n=10,
    p=0.5,
    size=1000
)

print(np.mean(samples))
```

```python
import numpy as np

samples = np.random.normal(
    loc=100,
    scale=10,
    size=10000
)

print(np.mean(samples))
print(np.std(samples))
```
