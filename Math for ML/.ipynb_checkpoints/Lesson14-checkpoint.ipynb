```python
w = 10

lr = 0.1

for i in range(10):

    grad = 2*(w-5)

    w = w - lr*grad

    print(w)
```

```python
import torch

# Define the simplest possible model (a single linear layer)
# This model takes 10 inputs and produces 2 outputs
model = torch.nn.Linear(10, 2)

# Now model.parameters() exists and can be passed to the optimizer
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

```python
import torch
model = torch.nn.Linear(10, 2)
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```
