```python
import numpy as np

def sigmoid(x):
    return 1/(1+np.exp(-x))

x = np.array([-2,0,2])

print(sigmoid(x))
```

```python
import numpy as np

def relu(x):
    return np.maximum(0,x)

x = np.array([-2,0,2])

print(relu(x))
```
