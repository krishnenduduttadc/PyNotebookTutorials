```python
import numpy as np

X = np.array([[2,3]])

W = np.array([
    [0.5],
    [0.2]
])

b = 1

Z = X @ W + b

print(Z)
```

```python
import numpy as np

z = np.array([2,1,0])

softmax = np.exp(z)/np.sum(np.exp(z))

print(softmax)
```

```python
import numpy as np

x = 3

w = 2

b = -4

z = w*x + b

p = 1/(1+np.exp(-z))

print("Probability:", p)
```
