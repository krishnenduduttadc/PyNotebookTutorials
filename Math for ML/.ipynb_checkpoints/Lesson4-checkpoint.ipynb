```python
import numpy as np

A = np.array([
    [1,2],
    [3,4]
])

print(A)
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(A + B)
```

```python
A = np.array([[1,2],[3,4]])

B = np.array([[5,6],[7,8]])

print(A @ B)
```

```python
A = np.array([
    [1,2,3],
    [4,5,6]
])

print(A.T)
```

```python
import numpy as np

X = np.array([[1,2]])

W = np.array([
    [0.5,0.2],
    [0.3,0.7]
])

Z = X @ W

print(Z)
```
