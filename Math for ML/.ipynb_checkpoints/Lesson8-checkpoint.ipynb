```python
import numpy as np

samples = np.random.normal(
    loc=100,
    scale=10,
    size=10000
)

print(np.mean(samples))
print(np.std(samples))
```

```python
w = 10

lr = 0.1

for i in range(15):

    loss = (w-5)**2

    grad = 2*(w-5)

    w = w - lr*grad

    print(
        i,
        round(w,4),
        round(loss,4)
    )
```
