```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import Ridge

# 1. Create a dummy regression dataset
X, y = make_regression(n_samples=100, n_features=3, noise=10, random_state=42)

# --- Your original code ---
# 2. Initialize the model with a regularization strength of 1.0
model = Ridge(alpha=1.0)

# 3. Fit the model to the data
model.fit(X, y)
# --------------------------

# 4. Inspect the results
print("--- Ridge Regression Results ---")
print(f"Intercept (w0): {model.intercept_:.3f}")
print(f"Coefficients (weights): {np.round(model.coef_, 3)}")

# 5. Predict on new data
new_data = np.array([[0.5, -1.2, 0.1]])
prediction = model.predict(new_data)
print(f"Prediction for new data: {prediction[0]:.3f}")
```

```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import Lasso

# 1. Create a dataset with 6 features, but only 3 of them actually matter
X, y = make_regression(n_samples=100, n_features=6, n_informative=3, noise=5, random_state=42)

# --- Your original code ---
# 2. Initialize the model (alpha controls penalty strength)
model = Lasso(alpha=0.1)

# 3. Fit the model
model.fit(X, y)
# --------------------------

# 4. Inspect the results to see feature selection
print("--- Lasso Regression Results ---")
print(f"Intercept: {model.intercept_:.3f}")
print("Coefficients:")
for i, coef in enumerate(model.coef_):
    print(f"  Feature {i}: {coef:.3f}")
```

```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import ElasticNet

# 1. Create a dataset with correlated features
X, y = make_regression(n_samples=100, n_features=5, n_informative=3, noise=10, random_state=42)

# --- Your original code ---
# 2. Initialize the model
model = ElasticNet(
    alpha=0.1,    # Total regularization penalty strength
    l1_ratio=0.5  # Equal mix of Lasso and Ridge
)

# 3. Fit the model
model.fit(X, y)
# --------------------------

# 4. Inspect the resulting coefficients
print("--- ElasticNet Regression Results ---")
print(f"Intercept: {model.intercept_:.3f}")
print(f"Coefficients: {np.round(model.coef_, 3)}")
```

```python
import torch
import torch.nn as nn

# 1. Define the model architecture (Your original code)
model = nn.Sequential(
    nn.Linear(100, 50),  # Layer 1: 100 input features -> 50 hidden units
    nn.ReLU(),           # Activation: Introduces non-linearity
    nn.Dropout(0.5),     # Regularization: Randomly zeros 50% of the channels
    nn.Linear(50, 1)     # Layer 2: 50 hidden units -> 1 output prediction
)

# 2. Create sample input data 
# PyTorch expects (batch_size, num_features). Here: batch of 4 samples, 100 features each.
X = torch.randn(4, 100)

# 3. Put the model in evaluation mode (turns off Dropout)
model.eval()

# 4. Pass the data through the network
with torch.no_grad(): # Disables gradient calculation to save memory during inference
    predictions = model(X)

# 5. Check the results
print("--- PyTorch Model Execution ---")
print(f"Input shape:  {X.shape}")
print(f"Output shape: {predictions.shape}")
print(f"Predictions:\n{predictions}")
```
