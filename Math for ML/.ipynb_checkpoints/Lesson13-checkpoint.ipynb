```python
import numpy as np

p = np.array([0.5,0.5])

q = np.array([0.4,0.6])

kl = np.sum(
    p*np.log(p/q)
)

print(kl)
```

```python
from sklearn.decomposition import PCA
import numpy as np

X = np.array([
    [170,70],
    [175,75],
    [180,80]
])

pca = PCA(n_components=1)

X_new = pca.fit_transform(X)

print(X_new)
```

```python
import numpy as np

A = np.array([
    [1,2],
    [3,4]
])

U,S,Vt = np.linalg.svd(A)

print(U)
print(S)
print(Vt)
```
