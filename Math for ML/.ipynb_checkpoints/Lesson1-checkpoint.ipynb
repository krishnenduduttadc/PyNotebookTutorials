```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x^2")
plt.grid()
plt.show()
```

```python
def f(x):
    return x**2

x = 3

h = 0.0001

derivative = (f(x+h) - f(x))/h

print(derivative)
```
