```python
import numpy as np

A = np.array([1,2])

B = np.array([2,4])

cosine = np.dot(A,B) / (
    np.linalg.norm(A)
    * np.linalg.norm(B)
)

print(cosine)
```

```python
from gensim.models import Word2Vec

sentences = [
    ["i","love","machine","learning"],
    ["i","love","deep","learning"]
]

model = Word2Vec(
    sentences,
    vector_size=10,
    window=2,
    min_count=1
)

print(
    model.wv["learning"]
)
```
