```python
# pip install requests  # run once if you don't have requests
import requests
import json

URL = "http://localhost:11434/api/generate"   # Ollama generate endpoint
payload = {
    "model": "llama3",                # change to a model you have pulled
    "prompt": "Explain gradient descent in 3 simple sentences.",
    "stream": False                   # set True if you want streaming chunks
}

resp = requests.post(URL, json=payload, timeout=120)
resp.raise_for_status()
# The server may send a single JSON or streaming JSON lines depending on settings.
try:
    data = resp.json()
    print(json.dumps(data, indent=2)[:2000])   # print truncated for readability
except ValueError:
    # fallback: print raw text if not JSON
    print(resp.text[:2000])
```

```python
import requests
URL = "http://localhost:11434/api/generate"
payload = {"model": "llama3", "prompt": "Write a short poem about code.", "stream": True}

with requests.post(URL, json=payload, stream=True) as r:
    r.raise_for_status()
    for line in r.iter_lines(decode_unicode=True):
        if not line:
            continue
        # each line is usually JSON — parse and extract text
        try:
            obj = json.loads(line)
            # typical streaming object has 'output' or 'content' field depending on model
            print(obj)
        except Exception:
            print("non-json chunk:", line)
```
