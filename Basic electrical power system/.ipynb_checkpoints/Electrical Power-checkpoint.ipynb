```python
# ============================================
# OHM'S LAW
# V = I × R
# Interactive Circuit Visualization
# ============================================

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
import schemdraw
import schemdraw.elements as elm
from IPython.display import display, Math, Markdown

display(Markdown("# ⚡ Ohm's Law"))
display(Math(r"V = I \times R"))

def ohms_law(voltage=12.0, resistance=6.0):

    current = voltage / resistance
    power = voltage * current

    print(f"Voltage     : {voltage:.2f} V")
    print(f"Resistance  : {resistance:.2f} Ω")
    print(f"Current     : {current:.2f} A")
    print(f"Power       : {power:.2f} W")

    # -------------------------
    # Circuit Diagram
    # -------------------------
    d = schemdraw.Drawing()

    d += elm.SourceV().up().label(f'{voltage:.1f} V')
    d += elm.Line().right()
    d += elm.Resistor().right().label(f'{resistance:.1f} Ω')
    d += elm.Line().down()
    d += elm.Line().left()
    d += elm.Line().up()

    display(d.draw())

    # -------------------------
    # Current Visualization
    # -------------------------

    fig, ax = plt.subplots(figsize=(7,1.6))

    ax.plot([0,10],[0,0],linewidth=6)

    arrow_length = min(current,10)

    ax.arrow(
        1,
        0.1,
        arrow_length,
        0,
        head_width=0.08,
        head_length=0.25,
        length_includes_head=True
    )

    ax.text(5,0.3,f"Current = {current:.2f} A",fontsize=12,ha='center')

    ax.set_xlim(0,10)
    ax.set_ylim(-0.4,0.8)
    ax.axis('off')

    plt.show()

interact(
    ohms_law,
    voltage=FloatSlider(value=12,min=1,max=240,step=1,description='Voltage'),
    resistance=FloatSlider(value=6,min=1,max=100,step=1,description='Resistance')
);
```

```python
#DC Power (P = VI, I²R, V²/R)
```

```python
# ============================================
# DC POWER
# P = VI = I²R = V²/R
# Interactive Circuit Visualization
# ============================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from ipywidgets import interact, FloatSlider
import schemdraw
import schemdraw.elements as elm
from IPython.display import display, Math, Markdown

display(Markdown("# ⚡ DC Electrical Power"))
display(Math(r"P = VI = I^2R = \frac{V^2}{R}"))

def dc_power(voltage=24.0, resistance=12.0):

    current = voltage / resistance
    power1 = voltage * current
    power2 = current**2 * resistance
    power3 = voltage**2 / resistance

    print(f"Voltage      : {voltage:.2f} V")
    print(f"Resistance   : {resistance:.2f} Ω")
    print(f"Current      : {current:.2f} A")
    print(f"P = V×I      : {power1:.2f} W")
    print(f"P = I²R      : {power2:.2f} W")
    print(f"P = V²/R     : {power3:.2f} W")

    # -----------------------------------
    # Circuit Diagram
    # -----------------------------------

    d = schemdraw.Drawing()

    d += elm.SourceV().up().label(f"{voltage:.1f} V")
    d += elm.Line().right()
    d += elm.Resistor().right().label(f"{resistance:.1f} Ω")
    d += elm.Line().down()
    d += elm.Line().left()
    d += elm.Line().up()

    display(d.draw())

    # -----------------------------------
    # Power Meter Visualization
    # -----------------------------------

    fig, ax = plt.subplots(figsize=(8,2.5))

    max_power = 1000
    ratio = min(power1/max_power,1)

    ax.add_patch(Rectangle((0,0),10,1,
                           fill=False,
                           linewidth=2))

    ax.add_patch(Rectangle((0,0),10*ratio,1))

    ax.text(5,1.25,
            f"Load Power = {power1:.2f} W",
            ha='center',
            fontsize=13)

    if power1 < 100:
        status="🟢 LOW LOAD"
    elif power1 < 500:
        status="🟡 MEDIUM LOAD"
    else:
        status="🔴 HIGH LOAD"

    ax.text(5,-0.4,status,
            ha='center',
            fontsize=12)

    ax.set_xlim(0,10)
    ax.set_ylim(-0.8,1.8)
    ax.axis('off')

    plt.show()


interact(
    dc_power,
    voltage=FloatSlider(
        value=24,
        min=1,
        max=240,
        step=1,
        description='Voltage'
    ),
    resistance=FloatSlider(
        value=12,
        min=1,
        max=100,
        step=1,
        description='Resistance'
    )
);
```

```python
#Single-Phase AC Power & Power Factor
```

```python
# ==========================================================
# SINGLE PHASE AC POWER
# P = V I cosφ
# S = V I
# Q = V I sinφ
# Interactive Power Triangle
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Single Phase AC Power"))

display(Math(r"P=VI\cos\phi"))
display(Math(r"S=VI"))
display(Math(r"Q=VI\sin\phi"))

def ac_power(voltage=230,
             current=10,
             pf=0.8):

    # ----------------------------
    # Calculations
    # ----------------------------

    S = voltage * current              # VA
    P = S * pf                         # W
    phi = np.arccos(pf)
    Q = S * np.sin(phi)                # VAR

    print(f"Voltage          : {voltage:.1f} V")
    print(f"Current          : {current:.2f} A")
    print(f"Power Factor     : {pf:.2f}")
    print(f"Phase Angle      : {np.degrees(phi):.2f}°")
    print()
    print(f"Real Power (P)   : {P:.2f} W")
    print(f"Reactive (Q)     : {Q:.2f} VAR")
    print(f"Apparent (S)     : {S:.2f} VA")

    # ----------------------------
    # Power Triangle
    # ----------------------------

    fig, ax = plt.subplots(figsize=(7,7))

    ax.plot([0,P],[0,0],linewidth=3,label='Real Power (P)')
    ax.plot([P,P],[0,Q],linewidth=3,label='Reactive Power (Q)')
    ax.plot([0,P],[0,Q],linewidth=3,label='Apparent Power (S)')

    ax.scatter([0,P,P],[0,0,Q],s=60)

    ax.text(P/2,-S*0.03,"P",fontsize=13,ha='center')

    ax.text(P+S*0.02,Q/2,"Q",fontsize=13)

    ax.text(P/2,Q/2,"S",fontsize=13)

    theta=np.linspace(0,phi,100)

    r=min(P,Q)*0.35

    ax.plot(r*np.cos(theta),r*np.sin(theta))

    ax.text(r*1.1*np.cos(phi/2),
            r*1.1*np.sin(phi/2),
            f"φ={np.degrees(phi):.1f}°")

    ax.set_title("Power Triangle")

    ax.set_xlabel("Real Power (W)")
    ax.set_ylabel("Reactive Power (VAR)")

    ax.grid(True)

    ax.set_aspect('equal')

    plt.show()

    # ----------------------------
    # Power Meter
    # ----------------------------

    fig, ax = plt.subplots(figsize=(8,2))

    ax.barh(["Real"],[P])
    ax.barh(["Reactive"],[Q])
    ax.barh(["Apparent"],[S])

    ax.set_xlabel("Power")

    plt.show()


interact(
    ac_power,

    voltage=FloatSlider(
        value=230,
        min=100,
        max=260,
        step=5,
        description='Voltage'
    ),

    current=FloatSlider(
        value=10,
        min=1,
        max=100,
        step=1,
        description='Current'
    ),

    pf=FloatSlider(
        value=0.8,
        min=0.2,
        max=1.0,
        step=0.01,
        description='PF'
    )
);
```

```python
#Three-Phase Power (Interactive RYB Phasor & Power Calculation)
```

```python
# ==========================================================
# THREE-PHASE POWER
# P = √3 V I PF
# Interactive Phasor Diagram
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown, Math
import schemdraw
import schemdraw.elements as elm

display(Markdown("# ⚡ Three-Phase AC Power"))

display(Math(r"P=\sqrt3VI\cos\phi"))
display(Math(r"S=\sqrt3VI"))
display(Math(r"Q=\sqrt3VI\sin\phi"))

SQRT3 = np.sqrt(3)

def three_phase(voltage=415,
                current=50,
                pf=0.85):

    # ----------------------------
    # Calculations
    # ----------------------------

    S = SQRT3 * voltage * current
    P = S * pf

    phi = np.arccos(pf)
    Q = S * np.sin(phi)

    print(f"Line Voltage     : {voltage:.1f} V")
    print(f"Line Current     : {current:.1f} A")
    print(f"Power Factor     : {pf:.2f}")
    print(f"Phase Angle      : {np.degrees(phi):.2f}°")
    print()

    print(f"Real Power       : {P/1000:.2f} kW")
    print(f"Reactive Power   : {Q/1000:.2f} kVAR")
    print(f"Apparent Power   : {S/1000:.2f} kVA")

    # ===================================================
    # Three-Phase Circuit
    # ===================================================

    d = schemdraw.Drawing()

    d += elm.SourceSin().up().label("R")
    d += elm.Line().right()

    d.push()

    d += elm.Resistor().right().label("Load")

    d.pop()

    d += elm.Line().down()

    d += elm.SourceSin().up().label("Y")

    d += elm.Line().down()

    d += elm.SourceSin().up().label("B")

    display(d.draw())

    # ===================================================
    # RYB Phasor Diagram
    # ===================================================

    fig, ax = plt.subplots(figsize=(7,7))

    radius = 1

    ax.add_patch(Circle((0,0),radius,
                        fill=False,
                        linestyle='--'))

    angles = [0,-120,120]
    labels = ['R','Y','B']

    for ang,label in zip(angles,labels):

        rad = np.radians(ang)

        x = np.cos(rad)
        y = np.sin(rad)

        ax.arrow(0,0,
                 x*0.9,
                 y*0.9,
                 head_width=0.05,
                 length_includes_head=True)

        ax.text(x*1.05,
                y*1.05,
                label,
                fontsize=14,
                weight='bold')

    ax.set_xlim(-1.3,1.3)
    ax.set_ylim(-1.3,1.3)

    ax.set_aspect('equal')
    ax.grid(True)

    ax.set_title("Three-Phase RYB Phasor")

    plt.show()

    # ===================================================
    # Power Triangle
    # ===================================================

    fig, ax = plt.subplots(figsize=(7,6))

    ax.plot([0,P],[0,0],lw=3)
    ax.plot([P,P],[0,Q],lw=3)
    ax.plot([0,P],[0,Q],lw=3)

    ax.text(P/2,-S*0.03,"P")

    ax.text(P+S*0.02,Q/2,"Q")

    ax.text(P/2,Q/2,"S")

    theta=np.linspace(0,phi,100)

    r=min(P,Q)*0.35

    ax.plot(r*np.cos(theta),
            r*np.sin(theta))

    ax.text(r*1.2*np.cos(phi/2),
            r*1.2*np.sin(phi/2),
            f"{np.degrees(phi):.1f}°")

    ax.set_xlabel("Real Power (W)")
    ax.set_ylabel("Reactive Power (VAR)")
    ax.grid(True)

    plt.show()

    # ===================================================
    # Summary
    # ===================================================

    fig,ax=plt.subplots(figsize=(8,2))

    values=[P/1000,Q/1000,S/1000]

    ax.bar(["kW","kVAR","kVA"],values)

    ax.set_ylabel("Power")

    plt.show()


interact(

    three_phase,

    voltage=FloatSlider(
        value=415,
        min=380,
        max=11000,
        step=5,
        description="Voltage"
    ),

    current=FloatSlider(
        value=50,
        min=1,
        max=500,
        step=1,
        description="Current"
    ),

    pf=FloatSlider(
        value=0.85,
        min=0.20,
        max=1.00,
        step=0.01,
        description="PF"
    )

);
```

```python
# ==========================================================
# POWER FACTOR CORRECTION
# Capacitor Bank Effect
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Arc
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Power Factor Improvement"))

display(Math(r"PF=\cos\phi"))
display(Math(r"Q_c=P(\tan\phi_1-\tan\phi_2)"))

SQRT3 = np.sqrt(3)

def pf_correction(power_kw=100,
                  voltage=415,
                  pf_before=0.70,
                  pf_after=0.95):

    P = power_kw * 1000

    phi1 = np.arccos(pf_before)
    phi2 = np.arccos(pf_after)

    Q1 = P * np.tan(phi1)
    Q2 = P * np.tan(phi2)

    capacitor = Q1 - Q2

    S1 = P / pf_before
    S2 = P / pf_after

    I_before = S1 / (SQRT3 * voltage)
    I_after  = S2 / (SQRT3 * voltage)

    print("=========== BEFORE ===========")
    print(f"Power Factor : {pf_before:.2f}")
    print(f"Current      : {I_before:.2f} A")
    print(f"kVA          : {S1/1000:.2f}")
    print(f"kVAR         : {Q1/1000:.2f}")

    print()

    print("=========== AFTER ===========")
    print(f"Power Factor : {pf_after:.2f}")
    print(f"Current      : {I_after:.2f} A")
    print(f"kVA          : {S2/1000:.2f}")
    print(f"kVAR         : {Q2/1000:.2f}")

    print()

    print(f"Required Capacitor = {capacitor/1000:.2f} kVAR")

    print(f"Current Reduction  = {I_before-I_after:.2f} A")

    # -----------------------------------------
    # Power Triangle
    # -----------------------------------------

    fig,ax=plt.subplots(figsize=(7,7))

    # Before

    ax.plot([0,P],[0,0],lw=3,label="Real Power")

    ax.plot([P,P],[0,Q1],lw=3,label="Reactive Before")

    ax.plot([0,P],[0,Q1],lw=3,label="Apparent Before")

    # After

    ax.plot([P,P],[0,Q2],lw=3)

    ax.plot([0,P],[0,Q2],lw=3,label="Apparent After")

    ax.text(P/2,-P*0.03,"P")

    ax.text(P+P*0.02,Q1/2,"Q")

    ax.set_xlabel("Real Power (W)")
    ax.set_ylabel("Reactive Power (VAR)")
    ax.set_title("Power Triangle Before & After PF Correction")

    ax.grid(True)

    plt.legend()

    plt.show()

    # -----------------------------------------
    # Current Comparison
    # -----------------------------------------

    fig,ax=plt.subplots(figsize=(6,4))

    currents=[I_before,I_after]

    labels=["Before","After"]

    ax.bar(labels,currents)

    ax.set_ylabel("Current (A)")
    ax.set_title("Current Reduction")

    plt.show()

    # -----------------------------------------
    # kVAR Comparison
    # -----------------------------------------

    fig,ax=plt.subplots(figsize=(6,4))

    kvar=[Q1/1000,Q2/1000]

    ax.bar(["Before","After"],kvar)

    ax.set_ylabel("Reactive Power (kVAR)")
    ax.set_title("Reactive Power Reduction")

    plt.show()

interact(

    pf_correction,

    power_kw=FloatSlider(
        value=100,
        min=10,
        max=1000,
        step=10,
        description="kW"
    ),

    voltage=FloatSlider(
        value=415,
        min=380,
        max=11000,
        step=5,
        description="Voltage"
    ),

    pf_before=FloatSlider(
        value=0.70,
        min=0.40,
        max=0.95,
        step=0.01,
        description="PF Before"
    ),

    pf_after=FloatSlider(
        value=0.95,
        min=0.70,
        max=1.00,
        step=0.01,
        description="PF After"
    )

);
```

```python
#Ideal Transformer (Turns Ratio & Voltage Transformation)
```

```python
# ==========================================================
# IDEAL TRANSFORMER
# Vp/Vs = Np/Ns
# Interactive Step-Up / Step-Down Transformer
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Arc
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown, Math
import schemdraw
import schemdraw.elements as elm

display(Markdown("# ⚡ Ideal Transformer"))

display(Math(r"\frac{V_p}{V_s}=\frac{N_p}{N_s}"))

display(Math(r"\frac{I_p}{I_s}=\frac{N_s}{N_p}"))

def transformer(primary_voltage=11000,
                primary_turns=500,
                secondary_turns=19,
                load_kVA=100):

    # -----------------------------
    # Calculations
    # -----------------------------

    ratio = secondary_turns / primary_turns

    secondary_voltage = primary_voltage * ratio

    primary_current = load_kVA*1000/primary_voltage

    secondary_current = load_kVA*1000/secondary_voltage

    print("========== Transformer ==========")

    print(f"Primary Voltage   : {primary_voltage:.1f} V")
    print(f"Primary Turns     : {primary_turns:.0f}")

    print()

    print(f"Secondary Turns   : {secondary_turns:.0f}")
    print(f"Secondary Voltage : {secondary_voltage:.1f} V")

    print()

    print(f"Rating            : {load_kVA:.1f} kVA")

    print()

    print(f"Primary Current   : {primary_current:.2f} A")
    print(f"Secondary Current : {secondary_current:.2f} A")

    print()

    if secondary_voltage > primary_voltage:
        print("Operation : STEP-UP Transformer")
    elif secondary_voltage < primary_voltage:
        print("Operation : STEP-DOWN Transformer")
    else:
        print("Operation : 1:1 Isolation Transformer")

    # ---------------------------------
    # Transformer Circuit
    # ---------------------------------

    d = schemdraw.Drawing()

    d += elm.SourceSin().up().label(f"{primary_voltage:.0f} V")

    d += elm.Line().right()

    d += elm.Transformer().label("Ideal Transformer")

    d += elm.Line().right()

    d += elm.Resistor().down().label("Load")

    display(d.draw())

    # ---------------------------------
    # Coil Visualization
    # ---------------------------------

    fig,ax = plt.subplots(figsize=(8,5))

    # Primary

    for i in range(int(primary_turns/25)):
        y=i*0.12
        ax.add_patch(
            Arc((2,y),0.4,0.18,
                theta1=0,
                theta2=180)
        )

    # Secondary

    for i in range(int(secondary_turns/2)):
        y=i*0.12
        ax.add_patch(
            Arc((5,y),0.4,0.18,
                theta1=180,
                theta2=360)
        )

    ax.plot([3.2,3.2],[0,3],'k--',linewidth=2)

    ax.text(1.5,3.2,"Primary")

    ax.text(4.5,3.2,"Secondary")

    ax.set_xlim(0,7)
    ax.set_ylim(-0.2,3.8)

    ax.set_aspect('equal')
    ax.axis('off')

    plt.title("Transformer Windings")

    plt.show()

    # ---------------------------------
    # Voltage & Current Comparison
    # ---------------------------------

    fig,ax=plt.subplots(figsize=(8,4))

    labels=["Voltage","Current"]

    primary=[primary_voltage,primary_current]

    secondary=[secondary_voltage,secondary_current]

    x=np.arange(len(labels))
    w=0.35

    ax.bar(x-w/2,primary,w,label="Primary")

    ax.bar(x+w/2,secondary,w,label="Secondary")

    ax.set_xticks(x)

    ax.set_xticklabels(labels)

    ax.legend()

    plt.title("Primary vs Secondary")

    plt.show()

interact(

    transformer,

    primary_voltage=FloatSlider(
        value=11000,
        min=230,
        max=33000,
        step=230,
        description="Vp"
    ),

    primary_turns=FloatSlider(
        value=500,
        min=50,
        max=1000,
        step=10,
        description="Np"
    ),

    secondary_turns=FloatSlider(
        value=19,
        min=5,
        max=1000,
        step=1,
        description="Ns"
    ),

    load_kVA=FloatSlider(
        value=100,
        min=10,
        max=5000,
        step=10,
        description="kVA"
    )

);
```

```python
# ==========================================================
# TRANSFORMER FULL LOAD CURRENT
# Single Phase & Three Phase
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from ipywidgets import interact, FloatSlider, Dropdown
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Transformer Full Load Current"))

display(Math(r"I_{1\phi}=\frac{kVA\times1000}{V}"))
display(Math(r"I_{3\phi}=\frac{kVA\times1000}{\sqrt3\times V}"))

SQRT3 = np.sqrt(3)

def transformer_loading(kva=250,
                        voltage=415,
                        phase='Three Phase',
                        loading=80):

    # ------------------------------
    # Full Load Current
    # ------------------------------

    if phase == 'Single Phase':
        full_current = (kva * 1000) / voltage
    else:
        full_current = (kva * 1000) / (SQRT3 * voltage)

    actual_current = full_current * loading / 100

    actual_kva = kva * loading / 100

    print("========== Transformer ==========\n")

    print(f"Rating            : {kva:.1f} kVA")
    print(f"Voltage           : {voltage:.1f} V")
    print(f"System            : {phase}")
    print()

    print(f"Full Load Current : {full_current:.2f} A")
    print(f"Operating Load    : {loading:.1f}%")
    print(f"Actual Current    : {actual_current:.2f} A")
    print(f"Actual Loading    : {actual_kva:.1f} kVA")

    print()

    if loading < 60:
        status = "🟢 Lightly Loaded"
    elif loading < 90:
        status = "🟡 Normal Loading"
    elif loading <= 100:
        status = "🟠 Near Full Load"
    else:
        status = "🔴 Overloaded"

    print(status)

    # -----------------------------------
    # Loading Meter
    # -----------------------------------

    fig, ax = plt.subplots(figsize=(8,2))

    ax.add_patch(Rectangle((0,0),100,1,
                           fill=False,
                           linewidth=2))

    color = "green"

    if loading > 60:
        color = "gold"

    if loading > 90:
        color = "orange"

    if loading > 100:
        color = "red"

    ax.add_patch(
        Rectangle((0,0),
                  min(loading,100),
                  1,
                  color=color)
    )

    ax.text(50,1.25,
            f"{loading:.1f}% Loaded",
            ha='center',
            fontsize=13)

    ax.set_xlim(0,110)
    ax.set_ylim(-0.3,1.6)

    ax.axis("off")

    plt.show()

    # -----------------------------------
    # Current Gauge
    # -----------------------------------

    fig, ax = plt.subplots(figsize=(6,4))

    ax.bar(["Full Load","Actual"],
           [full_current,actual_current])

    ax.set_ylabel("Current (A)")
    ax.set_title("Transformer Current")

    plt.show()

    # -----------------------------------
    # kVA Comparison
    # -----------------------------------

    fig, ax = plt.subplots(figsize=(6,4))

    ax.bar(["Rated","Operating"],
           [kva,actual_kva])

    ax.set_ylabel("kVA")

    plt.title("Transformer Loading")

    plt.show()

interact(

    transformer_loading,

    kva=FloatSlider(
        value=250,
        min=25,
        max=5000,
        step=25,
        description="kVA"
    ),

    voltage=FloatSlider(
        value=415,
        min=230,
        max=33000,
        step=5,
        description="Voltage"
    ),

    phase=Dropdown(
        options=["Single Phase","Three Phase"],
        value="Three Phase",
        description="System"
    ),

    loading=FloatSlider(
        value=80,
        min=10,
        max=150,
        step=5,
        description="Loading %"
    )

);
```

```python
#Voltage Drop in Feeders (Single & Three Phase)
```

```python
# ==========================================================
# VOLTAGE DROP IN FEEDERS
#
# Single Phase:
# VD = 2 × I × R × L
#
# Three Phase:
# VD = √3 × I × R × L
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Dropdown
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Voltage Drop in Distribution Feeders"))

display(Math(r"VD_{1\phi}=2IRL"))
display(Math(r"VD_{3\phi}=\sqrt3IRL"))

SQRT3 = np.sqrt(3)

def feeder_voltage_drop(
    source_voltage=415,
    current=120,
    resistance=0.15,      # Ω/km
    length=2.5,           # km
    system="Three Phase"
):

    # ------------------------------------
    # Voltage Drop Calculation
    # ------------------------------------

    if system == "Single Phase":
        vd = 2 * current * resistance * length
    else:
        vd = SQRT3 * current * resistance * length

    receiving_voltage = source_voltage - vd
    percent_drop = vd/source_voltage*100

    print("========== FEEDER ==========\n")

    print(f"System             : {system}")
    print(f"Source Voltage     : {source_voltage:.1f} V")
    print(f"Current            : {current:.1f} A")
    print(f"Cable Resistance   : {resistance:.3f} Ω/km")
    print(f"Length             : {length:.2f} km")

    print()

    print(f"Voltage Drop       : {vd:.2f} V")
    print(f"Receiving Voltage  : {receiving_voltage:.2f} V")
    print(f"Voltage Drop (%)   : {percent_drop:.2f} %")

    print()

    if percent_drop < 3:
        status = "🟢 Excellent"
    elif percent_drop < 5:
        status = "🟡 Acceptable"
    else:
        status = "🔴 Excessive Voltage Drop"

    print(status)

    # ====================================
    # Voltage Profile Along Feeder
    # ====================================

    distance = np.linspace(0, length, 100)

    voltage = source_voltage - (vd/length)*distance

    plt.figure(figsize=(8,4))
    plt.plot(distance, voltage, linewidth=3)

    plt.title("Voltage Profile Along Feeder")

    plt.xlabel("Distance (km)")
    plt.ylabel("Voltage (V)")

    plt.grid(True)
    plt.show()

    # ====================================
    # Feeder Diagram
    # ====================================

    fig, ax = plt.subplots(figsize=(10,2))

    ax.plot([0,10],[0,0],linewidth=5)

    ax.scatter([0],[0],s=150)

    ax.scatter([10],[0],s=150)

    ax.text(0,-0.4,"Substation",ha="center")

    ax.text(10,-0.4,"Consumer",ha="center")

    ax.text(
        5,
        0.35,
        f"{length:.2f} km Feeder",
        ha='center',
        fontsize=12
    )

    ax.text(
        10,
        0.55,
        f"{receiving_voltage:.1f} V",
        color="blue",
        fontsize=12
    )

    ax.axis("off")

    plt.show()

    # ====================================
    # Voltage Comparison
    # ====================================

    plt.figure(figsize=(6,4))

    plt.bar(
        ["Source","Receiving"],
        [source_voltage,receiving_voltage]
    )

    plt.ylabel("Voltage (V)")
    plt.title("Voltage Comparison")

    plt.show()

interact(

    feeder_voltage_drop,

    source_voltage=FloatSlider(
        value=415,
        min=230,
        max=11000,
        step=5,
        description="Voltage"
    ),

    current=FloatSlider(
        value=120,
        min=10,
        max=600,
        step=5,
        description="Current"
    ),

    resistance=FloatSlider(
        value=0.15,
        min=0.02,
        max=1.00,
        step=0.01,
        description="Ω/km"
    ),

    length=FloatSlider(
        value=2.5,
        min=0.1,
        max=20,
        step=0.1,
        description="Length"
    ),

    system=Dropdown(
        options=["Single Phase","Three Phase"],
        value="Three Phase",
        description="System"
    )

);
```

```python
#Copper Loss (I²R Loss)
```

```python
# ==========================================================
# COPPER LOSS (I²R LOSS)
#
# Formula:
# Copper Loss = I²R
#
# Used in:
# • Feeders
# • Transformers
# • Cables
# • Busbars
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Copper Loss (I²R Loss)"))

display(Math(r"P_{loss}=I^2R"))

def copper_loss(current=150,
                resistance=0.25):

    # ----------------------------------
    # Calculations
    # ----------------------------------

    loss = current**2 * resistance

    print("========== COPPER LOSS ==========\n")

    print(f"Current           : {current:.2f} A")
    print(f"Resistance        : {resistance:.3f} Ω")

    print()

    print(f"Copper Loss       : {loss:.2f} W")
    print(f"Copper Loss       : {loss/1000:.2f} kW")

    # ----------------------------------
    # Heat Classification
    # ----------------------------------

    if loss < 1000:
        status = "🟢 Low Heating"

    elif loss < 5000:
        status = "🟡 Moderate Heating"

    elif loss < 15000:
        status = "🟠 High Heating"

    else:
        status = "🔴 Severe Heating"

    print(status)

    # ==================================
    # Current vs Copper Loss Curve
    # ==================================

    currents = np.linspace(0, current*1.2, 200)

    losses = currents**2 * resistance

    plt.figure(figsize=(8,5))

    plt.plot(currents, losses, linewidth=3)

    plt.scatter(current, loss, s=80)

    plt.title("Copper Loss Increases as Current²")

    plt.xlabel("Current (A)")
    plt.ylabel("Copper Loss (W)")

    plt.grid(True)

    plt.show()

    # ==================================
    # Heating Meter
    # ==================================

    fig, ax = plt.subplots(figsize=(8,2))

    max_loss = 20000

    percent = min(loss/max_loss,1)

    color = "green"

    if percent > 0.30:
        color = "gold"

    if percent > 0.60:
        color = "orange"

    if percent > 0.85:
        color = "red"

    ax.add_patch(Rectangle((0,0),100,1,
                           fill=False,
                           linewidth=2))

    ax.add_patch(Rectangle((0,0),
                           percent*100,
                           1,
                           color=color))

    ax.text(50,
            1.2,
            f"{loss:.0f} W Heat Generated",
            ha='center',
            fontsize=13)

    ax.axis("off")

    plt.show()

    # ==================================
    # Power Dissipation Example
    # ==================================

    resistor = np.ones((20,20))

    temperature = min(loss/15000,1)

    plt.figure(figsize=(3,3))

    plt.imshow(resistor,
               cmap='hot',
               vmin=0,
               vmax=1)

    plt.title("Conductor Heating")

    plt.axis("off")

    plt.show()

interact(

    copper_loss,

    current=FloatSlider(
        value=150,
        min=10,
        max=500,
        step=5,
        description="Current"
    ),

    resistance=FloatSlider(
        value=0.25,
        min=0.01,
        max=2,
        step=0.01,
        description="Resistance"
    )

);
```

```python
#Star (Y) & Delta (Δ) Connections
```

```python
# ==========================================================
# STAR (Y) & DELTA (Δ) CONNECTION
#
# Star:
# VL = √3 VP
# IL = IP
#
# Delta:
# VL = VP
# IL = √3 IP
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from ipywidgets import interact, FloatSlider, Dropdown
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Star (Y) and Delta (Δ) Connections"))

display(Math(r"Star:\;V_L=\sqrt3V_P,\qquad I_L=I_P"))
display(Math(r"Delta:\;V_L=V_P,\qquad I_L=\sqrt3I_P"))

SQRT3 = np.sqrt(3)

def star_delta(phase_voltage=240,
               phase_current=20,
               connection="Star (Y)"):

    # ------------------------------------
    # Calculations
    # ------------------------------------

    if connection == "Star (Y)":
        line_voltage = SQRT3 * phase_voltage
        line_current = phase_current

    else:
        line_voltage = phase_voltage
        line_current = SQRT3 * phase_current

    print("========== THREE PHASE ==========\n")

    print(f"Connection      : {connection}")
    print(f"Phase Voltage   : {phase_voltage:.2f} V")
    print(f"Phase Current   : {phase_current:.2f} A")

    print()

    print(f"Line Voltage    : {line_voltage:.2f} V")
    print(f"Line Current    : {line_current:.2f} A")

    # =====================================
    # Diagram
    # =====================================

    fig, ax = plt.subplots(figsize=(6,6))

    if connection == "Star (Y)":

        center = np.array([0,0])

        angles = [90,-30,210]

        labels = ['R','Y','B']

        for ang,label in zip(angles,labels):

            x = np.cos(np.radians(ang))
            y = np.sin(np.radians(ang))

            ax.plot([0,x],[0,y],linewidth=3)

            ax.scatter(x,y,s=80)

            ax.text(x*1.15,
                    y*1.15,
                    label,
                    fontsize=14)

        ax.scatter(0,0,s=80)

        ax.text(0,-0.2,"N",fontsize=13)

        plt.title("Star (Y) Connection")

    else:

        pts = np.array([
            [0,1],
            [-0.87,-0.5],
            [0.87,-0.5]
        ])

        poly = Polygon(
            pts,
            fill=False,
            linewidth=3
        )

        ax.add_patch(poly)

        labels = ['R','Y','B']

        for p,l in zip(pts,labels):

            ax.scatter(p[0],p[1],s=80)

            ax.text(
                p[0]*1.15,
                p[1]*1.15,
                l,
                fontsize=14
            )

        plt.title("Delta (Δ) Connection")

    ax.set_xlim(-1.5,1.5)
    ax.set_ylim(-1.5,1.5)

    ax.set_aspect("equal")
    ax.grid(True)

    plt.show()

    # =====================================
    # Voltage Comparison
    # =====================================

    fig,ax = plt.subplots(figsize=(6,4))

    ax.bar(
        ["Phase","Line"],
        [phase_voltage,line_voltage]
    )

    ax.set_ylabel("Voltage (V)")
    ax.set_title("Voltage Comparison")

    plt.show()

    # =====================================
    # Current Comparison
    # =====================================

    fig,ax = plt.subplots(figsize=(6,4))

    ax.bar(
        ["Phase","Line"],
        [phase_current,line_current]
    )

    ax.set_ylabel("Current (A)")
    ax.set_title("Current Comparison")

    plt.show()

interact(

    star_delta,

    phase_voltage=FloatSlider(
        value=240,
        min=50,
        max=11000,
        step=10,
        description="Phase V"
    ),

    phase_current=FloatSlider(
        value=20,
        min=1,
        max=500,
        step=1,
        description="Phase I"
    ),

    connection=Dropdown(
        options=["Star (Y)","Delta (Δ)"],
        value="Star (Y)",
        description="Connection"
    )

);
```

```python
# ==========================================================
# ENERGY CONSUMPTION
#
# Energy (kWh) = Power (kW) × Time (hours)
#
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown, Math

display(Markdown("# ⚡ Energy Consumption & Electricity Bill"))

display(Math(r"Energy\;(kWh)=Power(kW)\times Time(hours)"))

def energy_meter(power_kw=5,
                 hours=8,
                 tariff=7):

    # ----------------------------------
    # Calculations
    # ----------------------------------

    energy = power_kw * hours
    bill = energy * tariff

    print("========== ENERGY METER ==========\n")

    print(f"Connected Load : {power_kw:.2f} kW")
    print(f"Operating Time : {hours:.2f} hours")

    print()

    print(f"Energy Used    : {energy:.2f} kWh")
    print(f"Tariff         : ₹{tariff:.2f}/kWh")
    print(f"Estimated Bill : ₹{bill:.2f}")

    # ----------------------------------
    # Energy Meter Dial
    # ----------------------------------

    fig, ax = plt.subplots(figsize=(5,5))

    circle = Circle((0,0),1,
                    fill=False,
                    linewidth=2)

    ax.add_patch(circle)

    angle = 270 - (energy % 100) * 3.6
    rad = np.radians(angle)

    ax.plot([0,0.8*np.cos(rad)],
            [0,0.8*np.sin(rad)],
            linewidth=3)

    ax.text(0,-1.25,
            f"{energy:.1f} kWh",
            ha="center",
            fontsize=14)

    ax.set_xlim(-1.3,1.3)
    ax.set_ylim(-1.3,1.3)
    ax.set_aspect("equal")
    ax.axis("off")

    plt.title("Energy Meter")

    plt.show()

    # ----------------------------------
    # Power vs Energy
    # ----------------------------------

    fig, ax = plt.subplots(figsize=(8,4))

    t = np.linspace(0,hours,100)
    cumulative = power_kw * t

    ax.plot(t,cumulative,
            linewidth=3)

    ax.set_xlabel("Time (hours)")
    ax.set_ylabel("Energy (kWh)")
    ax.set_title("Cumulative Energy Consumption")

    ax.grid(True)

    plt.show()

    # ----------------------------------
    # Cost Breakdown
    # ----------------------------------

    fig, ax = plt.subplots(figsize=(6,4))

    ax.bar(
        ["Energy","Bill"],
        [energy,bill]
    )

    ax.set_title("Energy & Cost")

    plt.show()

interact(

    energy_meter,

    power_kw=FloatSlider(
        value=5,
        min=0.5,
        max=100,
        step=0.5,
        description="Power"
    ),

    hours=FloatSlider(
        value=8,
        min=1,
        max=24,
        step=1,
        description="Hours"
    ),

    tariff=FloatSlider(
        value=7,
        min=1,
        max=20,
        step=0.5,
        description="₹/kWh"
    )

);
```
