```python
import xml.etree.ElementTree as ET

# Example XML string (you can load from file instead)
xml_data = """
<root>
    <Id>734056032</Id>
    <Name>734056032</Name>
    <FromNode>734056032_P</FromNode>
    <ToNode>734056032_S</ToNode>
    <ServiceCenter>TRIAL_CNTR</ServiceCenter>
<PermitAreaId>STD2</PermitAreaId>
    <TfHasTertiaryWinding>false</TfHasTertiaryWinding>
    <TransformerModel>500_Three Phase GSU(1)_13.2_1</TransformerModel>
    <PrimPhase>ABC</PrimPhase>
    <SecPhase>ABCN</SecPhase>
    <PrimVolt>6.350852961</PrimVolt>
    <SecVolt>0.239600362</SecVolt>
    <Ground>2</Ground>
    <RGround>10</RGround>
    <XGround>0</XGround>
    <RGroundsec>10</RGroundsec>
    <XGroundsec>0</XGroundsec>
</root>
"""

# Parse XML
root = ET.fromstring(xml_data)

# Recursive function to print tags in order
def print_tags(element):
    print(element.tag)  # Print current tag
    for child in element:
        print_tags(child)  # Recursively print child tags

# Print tags from top to bottom
print_tags(root)
```
