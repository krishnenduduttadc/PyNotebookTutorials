```python
import pyautogui
import pywhatkit
import time
import os

# ====== SETTINGS ======
phone_number = "+919432957227"  # Include country code
image_path = "screenshot.png"
wait_time = 15  # Time to wait for WhatsApp Web to load
# =======================

# 1. Take Screenshot
print("Capturing screenshot...")
screenshot = pyautogui.screenshot()
screenshot.save(image_path)
print(f"Screenshot saved as {image_path}")

# 2. Wait a moment before sending
time.sleep(2)

# 3. Send Image via WhatsApp
print("Opening WhatsApp Web...")
pywhatkit.sendwhats_image(
    receiver=phone_number,
    img_path=image_path,
    caption="Here is the latest screenshot.",
    wait_time=wait_time
)

print("Screenshot sent successfully!")

# Optional: Remove screenshot after sending
# os.remove(image_path)
```

```python
import pyautogui
import smtplib
import ssl
from email.message import EmailMessage
import os
from datetime import datetime

# ====== CONFIGURATION ======
EMAIL_ADDRESS = "krishnenduduttadc@gmail.com"
APP_PASSWORD = "Aceprogrammer3005$"
SEND_TO = "meagain.krish@gmail.com"
# ===========================

def capture_screenshot():
    filename = f"screenshot_{datetime.now().strftime('%Y%m%d_%H%M%S')}.png"
    screenshot = pyautogui.screenshot()
    screenshot.save(filename)
    return filename

def send_email_with_attachment(file_path):
    msg = EmailMessage()
    msg["Subject"] = "Automated Screenshot"
    msg["From"] = EMAIL_ADDRESS
    msg["To"] = SEND_TO
    msg.set_content("Attached is the latest automated screenshot.")

    with open(file_path, "rb") as f:
        file_data = f.read()
        file_name = os.path.basename(file_path)

    msg.add_attachment(file_data, maintype="image", subtype="png", filename=file_name)

    context = ssl.create_default_context()

    with smtplib.SMTP_SSL("smtp.gmail.com", 465, context=context) as server:
        server.login(EMAIL_ADDRESS, APP_PASSWORD)
        server.send_message(msg)

# ====== RUN ======
screenshot_file = capture_screenshot()
send_email_with_attachment(screenshot_file)

print("Screenshot sent successfully!")
```
