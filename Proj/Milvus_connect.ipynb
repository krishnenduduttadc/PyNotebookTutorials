```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

canonicals = []
variants = []
embeddings = []

for c, v in data:
    canonicals.append(c)
    variants.append(v)
    embeddings.append(model.encode(v))
```

```python
collection.insert([canonicals, variants, embeddings])
```

```python
collection.create_index(
    field_name="embedding",
    index_params={
        "index_type": "AUTOINDEX",
        "metric_type": "L2",
        "params": {}
    }
)
```

```python
collection.load()
```

```python
def retrieve_canonical(query):
    query_vec = model.encode([query])

    results = collection.search(
        data=query_vec,
        anns_field="embedding",
        param={"metric_type": "L2"},
        limit=1,
        output_fields=["canonical", "variant"]
    )

    hit = results[0][0]
    return hit.entity.get("canonical")
```

```python
print(retrieve_canonical("Sii"))   # → S2
print(retrieve_canonical("S-one")) # → S1
print(retrieve_canonical("Si"))    # → S1
```

```python
def retrieve_canonical(query):
    query_vec = model.encode([query])

    results = collection.search(
        data=query_vec,
        anns_field="embedding",
        param={"metric_type": "L2"},
        limit=1,
        output_fields=["canonical", "variant"]
    )

    hit = results[0][0]
    return hit.entity.get("canonical")
```

```python
def fast_json(query):
    canonical = retrieve_canonical(query)
    return {canonical: True}
```

```python
print(fast_json("Sii"))        # → {'S2': True}
print(fast_json("section 3"))  # → {'S3': True}
print(fast_json("s-seven"))    # → {'S7': True}
print(fast_json("S_x"))        # → {'S10': True}
```
