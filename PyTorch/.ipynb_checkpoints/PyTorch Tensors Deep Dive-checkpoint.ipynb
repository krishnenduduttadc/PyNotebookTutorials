```python
import torch

# Basic tensors
a = torch.tensor([1, 2, 3])
b = torch.zeros((2, 3))
c = torch.ones((2, 2))
d = torch.randn((3, 3))

print(a, b, c, d)
```

```python
x = torch.randn(3, 4, requires_grad=True)

print("Shape:", x.shape)
print("Dtype:", x.dtype)
print("Device:", x.device)
print("Requires grad:", x.requires_grad)
```
