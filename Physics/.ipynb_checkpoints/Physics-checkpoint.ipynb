```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
```

```python
def motion(u=5, a=2):
    t = np.linspace(0, 10, 100)
    s = u*t + 0.5*a*t**2
    plt.plot(t, s)
    plt.xlabel("Time")
    plt.ylabel("Position")
    plt.grid()
    plt.show()

interact(motion, u=(-10,10), a=(-5,5))
```

```python
def force(m=1, a=5):
    print("Force =", m*a, "N")

interact(force, m=(1,10), a=(0,10))
```

```python
def kinetic_energy(m=1, v=5):
    print("KE =", 0.5*m*v**2)

interact(kinetic_energy, m=(1,10), v=(0,20))
```

```python
def rotation(omega=2):
    t = np.linspace(0, 10, 100)
    theta = omega*t
    x = np.cos(theta)
    y = np.sin(theta)
    plt.plot(x,y)
    plt.axis('equal')
    plt.show()

interact(rotation, omega=(0,10))
```

```python
def gravity(r=10):
    G = 6.67e-11
    M = 5.97e24
    m = 1
    F = G*M*m/r**2
    print("Force =", F)

interact(gravity, r=(1,100))
```

```python
def gas(T=300):
    n=1; R=8.314; V=1
    P = n*R*T/V
    print("Pressure =", P)

interact(gas, T=(100,1000))
```

```python
def heat(m=1, c=4200, dT=10):
    print("Heat =", m*c*dT)

interact(heat, m=(1,10), dT=(1,50))
```

```python
def shm(A=1, omega=2):
    t = np.linspace(0, 10, 100)
    x = A*np.cos(omega*t)
    plt.plot(t, x)
    plt.title("SHM")
    plt.grid()
    plt.show()

interact(shm, A=(1,5), omega=(1,10))
```

```python
def wave(k=1):
    x = np.linspace(0,10,100)
    y = np.sin(k*x)
    plt.plot(x,y)
    plt.title("Wave")
    plt.grid()
    plt.show()

interact(wave, k=(1,10))
```

```python
def lens(u=-20, f=10):
    v = 1/(1/f - 1/u)
    print("Image distance:", v)

interact(lens, u=(-50,-5), f=(5,20))
```

```python
def interference(d=1):
    x = np.linspace(0,10,1000)
    y = np.sin(x) + np.sin(x + d)
    plt.plot(x,y)
    plt.title("Interference")
    plt.show()

interact(interference, d=(0,5))
```

```python
def electrostatic(r=1):
    k=9e9; q1=1e-6; q2=1e-6
    F = k*q1*q2/r**2
    print("Force =", F)

interact(electrostatic, r=(0.1,10))
```

```python
def ohm(I=1, R=5):
    print("Voltage =", I*R)

interact(ohm, I=(0,10), R=(1,20))
```

```python
x, y = np.meshgrid(np.linspace(-5,5,20), np.linspace(-5,5,20))
B_x = -y/(x**2+y**2)
B_y = x/(x**2+y**2)

plt.quiver(x,y,B_x,B_y)
plt.title("Magnetic Field")
plt.show()
```

```python
def emf(rate=1):
    print("EMF ∝", rate)

interact(emf, rate=(1,10))
```

```python
def decay(l=0.5):
    t = np.linspace(0,10,100)
    N = np.exp(-l*t)
    plt.plot(t,N)
    plt.title("Radioactive Decay")
    plt.show()

interact(decay, l=(0.1,1))
```
