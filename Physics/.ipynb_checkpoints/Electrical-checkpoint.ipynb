```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def ohms_law_plot(voltage=5):
    R = np.linspace(1, 100, 100)
    I = voltage / R

    plt.figure()
    plt.plot(R, I)
    plt.xlabel("Resistance (Ohms)")
    plt.ylabel("Current (Amps)")
    plt.title(f"Ohm's Law (V = {voltage}V)")
    plt.grid()
    plt.show()

slider = widgets.FloatSlider(min=1, max=20, step=1, value=5, description='Voltage')

widgets.interactive(ohms_law_plot, voltage=slider)
```

```python
def rc_charging(R=1000, C=0.001, V0=5):
    t = np.linspace(0, 5, 100)
    tau = R * C
    V = V0 * (1 - np.exp(-t / tau))

    plt.figure()
    plt.plot(t, V)
    plt.xlabel("Time (seconds)")
    plt.ylabel("Voltage")
    plt.title(f"RC Charging Curve (R={R}, C={C})")
    plt.grid()
    plt.show()

widgets.interactive(
    rc_charging,
    R=widgets.IntSlider(min=100, max=5000, step=100, value=1000),
    C=widgets.FloatSlider(min=0.0001, max=0.01, step=0.0001, value=0.001),
    V0=widgets.FloatSlider(min=1, max=10, step=1, value=5)
)
```

```python
def ac_signal(freq=1, amplitude=1):
    t = np.linspace(0, 2, 500)
    y = amplitude * np.sin(2 * np.pi * freq * t)

    plt.figure()
    plt.plot(t, y)
    plt.xlabel("Time")
    plt.ylabel("Voltage")
    plt.title(f"AC Signal (f={freq} Hz)")
    plt.grid()
    plt.show()

widgets.interactive(
    ac_signal,
    freq=widgets.FloatSlider(min=0.5, max=10, step=0.5, value=1),
    amplitude=widgets.FloatSlider(min=1, max=10, step=1, value=1)
)

R_slider = widgets.IntSlider(min=100, max=5000, step=100, value=1000, description='Resistance')
C_slider = widgets.FloatSlider(min=0.0001, max=0.01, step=0.0001, value=0.001, description='Capacitance')

ui = widgets.VBox([R_slider, C_slider])

out = widgets.interactive_output(rc_charging, {'R': R_slider, 'C': C_slider})

display(ui, out)
```
