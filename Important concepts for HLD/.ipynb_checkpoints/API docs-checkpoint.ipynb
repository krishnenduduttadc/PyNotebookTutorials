SOAP API (XML-based Web Services)

```python
import shutil, subprocess, os
print("sys.executable:", __import__("sys").executable)
print("shutil.which('dot'):", shutil.which("dot"))

# Run dot -V and capture output
try:
    r = subprocess.run(["dot", "-V"], capture_output=True, text=True, check=True)
    print("dot -V stdout:", r.stdout)
    print("dot -V stderr:", r.stderr)
except subprocess.CalledProcessError as e:
    print("dot returned non-zero. stdout:", e.stdout)
    print("stderr:", e.stderr)
except FileNotFoundError:
    print("dot not found by subprocess.")

p = os.environ.get("PATH","").split(os.pathsep)
candidates = []
for d in p:
    for name in ("dot.exe","dot.bat","dot.cmd"):
        path = os.path.join(d, name)
        if os.path.exists(path):
            candidates.append(path)
print("First few PATH dirs checked:", p[-6:])
print("Found dot candidates (first 10):", candidates[:10])

# replace 'soap_case.gv' if your file is named differently
import pathlib
gv = pathlib.Path("soap_case.gv")
if gv.exists():
    try:
        subprocess.run(["dot", "-Tpng", str(gv), "-o", "soap_case_test.png"], check=True)
        print("Rendered soap_case_test.png successfully.")
    except Exception as e:
        print("Rendering failed:", e)
else:
    print("No soap_case.gv found in working dir:", pathlib.Path().resolve())
```

```python
from diagrams import Diagram, Cluster
from diagrams.onprem.client import User
from diagrams.onprem.network import Internet
from diagrams.onprem.compute import Server
from diagrams.programming.language import Java
from diagrams.generic.database import SQL
from diagrams.custom import Custom
from IPython.display import Image

with Diagram("SOAP API - Payment System (Legacy Integration)", show=False, filename="soap_case", outformat="png"):
    client = User("E-commerce Frontend")

    with Cluster("Backend"):
        frontend = Server("Frontend App")
        auth = Server("Auth Service")

        with Cluster("Payment Gateway - SOAP"):
            wsdl = Custom("WSDL", "./icons/wsdl.png")
            soap_handler = Java("SOAP Server")
            legacy_db = SQL("Payment DB")

        frontend >> auth >> soap_handler
        soap_handler >> wsdl >> legacy_db

    client >> Internet("HTTPS/XML") >> frontend

Image("soap_case.png")
```

```python
import os, sys
from pathlib import Path

# ---------- configure conda native DLL search path ----------
# adjust this if your conda base is elsewhere
CONDA_BASE = r"C:\Users\KRISHNENDU\anaconda3"
LIBBIN = Path(CONDA_BASE) / "Library" / "bin"

if sys.version_info >= (3, 8):
    try:
        os.add_dll_directory(str(LIBBIN))
        print("Added DLL dir:", LIBBIN)
    except Exception as e:
        # fallback: prepend to PATH
        os.environ["PATH"] = str(LIBBIN) + os.pathsep + os.environ.get("PATH", "")
        print("os.add_dll_directory failed, prepended to PATH. Error:", e)
else:
    os.environ["PATH"] = str(LIBBIN) + os.pathsep + os.environ.get("PATH", "")
    print("Prepended Library\\bin to PATH for older Python")

# ---------- prepare output directory ----------
OUTDIR = Path.cwd() / "diagrams_out"
OUTDIR.mkdir(parents=True, exist_ok=True)
print("Output directory:", OUTDIR)

# ---------- imports for diagrams ----------
from diagrams import Diagram, Cluster
from diagrams.onprem.client import User
from diagrams.onprem.network import Internet
from diagrams.onprem.compute import Server
from diagrams.programming.language import Java
from diagrams.generic.database import SQL
from diagrams.custom import Custom
from IPython.display import Image, display

# ---------- resolve custom icon (use only if exists) ----------
rel_icon = Path("./icons/wsdl.png")
if rel_icon.exists():
    icon_path = str(rel_icon.resolve())
    print("Using custom icon:", icon_path)
else:
    # try an absolute path you may use; otherwise, don't use Custom
    abs_icon_guess = Path.home() / "icons" / "wsdl.png"
    if abs_icon_guess.exists():
        icon_path = str(abs_icon_guess.resolve())
        print("Using custom icon from home icons:", icon_path)
    else:
        icon_path = None
        print("Custom icon not found (checked ./icons/wsdl.png and ~/icons/wsdl.png). Will not use Custom().")

# ---------- build diagram, write to OUTDIR ---------- 
outfile = OUTDIR / "soap_case"
with Diagram(
    "SOAP API - Payment System (Legacy Integration)",
    show=False,
    filename=str(outfile),   # diagrams will create soap_case.gv and soap_case.png in OUTDIR
    outformat="png",
):
    client = User("E-commerce Frontend")

    with Cluster("Backend"):
        frontend = Server("Frontend App")
        auth = Server("Auth Service")

        with Cluster("Payment Gateway - SOAP"):
            if icon_path:
                wsdl = Custom("WSDL", icon_path)
            else:
                # fallback to a simple labeled node if icon missing
                wsdl = Server("WSDL (no icon)")
            soap_handler = Java("SOAP Server")
            legacy_db = SQL("Payment DB")

        frontend >> auth >> soap_handler
        soap_handler >> wsdl >> legacy_db

    client >> Internet("HTTPS/XML") >> frontend

# ---------- show resulting PNG ----------
png_path = outfile.with_suffix(".png")
if png_path.exists():
    display(Image(str(png_path)))
else:
    print("Diagram created but PNG not found at:", png_path)
    print("Check for soap_case.gv in", OUTDIR)
```

 REST API (JSON over HTTP)

```python
from diagrams import Diagram, Cluster
from diagrams.onprem.client import User
from diagrams.onprem.compute import Server
from diagrams.generic.database import SQL
from diagrams.onprem.network import Internet
from IPython.display import Image

with Diagram("REST API - Product and Order Microservices", show=False, filename="rest_case", outformat="png"):
    client = User("React Frontend")

    with Cluster("Backend REST APIs"):
        auth = Server("Auth Service\n(/login, /signup)")
        product = Server("Product Service\n(/products)")
        order = Server("Order Service\n(/orders)")
        inventory = SQL("Inventory DB")

        product >> inventory
        order >> inventory

    client >> Internet("HTTPS/JSON") >> [auth, product, order]

Image("rest_case.png")
```

GraphQL API

```python
from diagrams import Diagram, Cluster
from diagrams.programming.framework import React
from diagrams.onprem.compute import Server
from diagrams.onprem.network import Internet
from diagrams.generic.database import SQL
from IPython.display import Image

with Diagram("GraphQL API Gateway - Unified Frontend Access", show=False, filename="graphql_case", outformat="png"):
    frontend = React("Next.js Frontend")

    with Cluster("API Gateway"):
        graphql = Server("GraphQL Gateway")

        with Cluster("Microservices"):
            product = Server("Product Service")
            order = Server("Order Service")
            auth = Server("Auth Service")

        graphql >> [product, order, auth]

    product >> SQL("Product DB")
    order >> SQL("Order DB")

    frontend >> Internet("HTTPS") >> graphql

Image("graphql_case.png")
```

gRPC API (Protocol Buffers over HTTP/2)

```python
from diagrams import Diagram, Cluster
from diagrams.onprem.compute import Server
from diagrams.generic.database import SQL
from diagrams.programming.language import Go
from diagrams.programming.language import Python
from IPython.display import Image

with Diagram("gRPC API - Internal Microservices", show=False, filename="grpc_case", outformat="png"):
    with Cluster("Backend System"):
        order_service = Go("Order Service")
        inventory_service = Python("Inventory Service")
        payment_service = Server("Payment Service")

        order_service >> inventory_service
        order_service >> payment_service

        inventory_service >> SQL("Inventory DB")
        payment_service >> SQL("Transaction DB")

Image("grpc_case.png")
```
