```python

# grpc_arch_diagram.py - paste this into a Jupyter cell and run (requires diagrams + graphviz)
from diagrams import Diagram, Cluster
from diagrams.onprem.client import User
from diagrams.onprem.network import Internet, Nginx
from diagrams.onprem.compute import Server

with Diagram("gRPC architecture (simple)", show=False, filename="grpc_arch", outformat="png"):
    user = User("Client App")

    with Cluster("Edge"):
        lb = Nginx("Envoy / Load-Balancer (optional)")
        proxy = Nginx("gRPC-Web Proxy (Envoy)")

    with Cluster("Backend Services (k8s/pods)"):
        svc_a = Server("Greeter Service\nPython/gRPC")
        svc_b = Server("Auth Service\nGo/gRPC")

    # HTTP/2 from client -> edge -> backend service
    user >> Internet("HTTP/2") >> lb >> proxy >> svc_a
    lb >> svc_b

```
