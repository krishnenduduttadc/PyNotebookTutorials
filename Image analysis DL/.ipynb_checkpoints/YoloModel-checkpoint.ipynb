```python
# !pip install ultralytics opencv-python Pillow


from ultralytics import YOLO
import cv2
from PIL import Image
from IPython.display import display

model = YOLO('yolov8n.pt')
image_path = r"E:\test.jpg"
results = model.predict(source=image_path, conf=0.25, iou=0.7, save=True)
print(f"--- Analysis Complete ---")


for r in results:
    names = r.names
    print("\nDetected Objects:")
    for box in r.boxes:
        # Extract coordinates and convert to list (xmin, ymin, xmax, ymax)
        coords = box.xyxy[0].tolist()
        # Get class ID and confidence score
        class_id = box.cls[0].item()
        conf = box.conf[0].item()
        # Get the class name
        class_name = names[int(class_id)]
        # Print the findings
        print(f" - {class_name} (Confidence: {conf:.2f}) at coordinates: [{int(coords[0])}, {int(coords[1])}, {int(coords[2])}, {int(coords[3])}]")


# Display the resulting image with bounding boxes (optional, but good for visualization)
print("\n--- Visualizing Result ---")
# The path to the saved image is typically in the 'save_dir' of the results object
saved_image_path = results[0].save_dir + '/' + image_path.split('/')[-1]

try:
    img_cv2 = cv2.imread(saved_image_path)
    img_rgb = cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB)

    # Display the image
    display(Image.fromarray(img_rgb))
except Exception as e:
    print(f"Could not load and display the saved image: {e}")
    print(f"Check the output directory: {results[0].save_dir}")
```
