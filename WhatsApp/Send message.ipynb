```python
import time
import openpyxl
import pywhatkit as kit

COUNTRY_CODE = "+91"

# Load Excel
workbook = openpyxl.load_workbook("messages.xlsx")
sheet = workbook.active

for row in range(2, sheet.max_row + 1):

    phone = sheet.cell(row=row, column=1).value
    message = sheet.cell(row=row, column=2).value

    if phone is None or message is None:
        continue

    # Convert to string and remove decimal if Excel stored it as float
    phone = str(phone).split(".")[0].strip()

    # Keep only digits
    phone = "".join(filter(str.isdigit, phone))

    # Must be 10 digits
    if len(phone) != 10:
        print(f"Skipping invalid number: {phone}")
        continue

    whatsapp_number = COUNTRY_CODE + phone

    try:
        print(f"Sending to {whatsapp_number}")

        kit.sendwhatmsg_instantly(
        phone_no=whatsapp_number,
        message=str(message),
        wait_time=25,      # was 15
        tab_close=True,
        close_time=5
        )

        print("✓ Sent")

    except Exception as e:



        
        print(f"✗ Failed: {e}")

    # Wait before sending the next message
    time.sleep(20)

print("All messages sent.")
```
