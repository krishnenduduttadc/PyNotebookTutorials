```python
import pandas as pd

# Small dataset: network traffic logs
data = {
    "src_ip": ["10.0.0.1", "10.0.0.2", "192.168.1.1"],
    "dst_ip": ["8.8.8.8", "8.8.4.4", "1.1.1.1"],
    "bytes": [120, 340, 560],
    "protocol": ["TCP", "UDP", "TCP"]
}

df = pd.DataFrame(data)

print("🔹 First rows:\n", df.head())
print("\n🔹 Describe stats:\n", df.describe())

# Filtering: TCP packets only
print("\n🔹 TCP traffic:\n", df[df["protocol"] == "TCP"])

# Grouping
print("\n🔹 Bytes per protocol:\n", df.groupby("protocol")["bytes"].sum())
```

```python
import pandas as pd

# Hardcoded log-like dataset
data = {
    "src_ip": ["10.0.0.1", "10.0.0.2", "192.168.1.10", "10.0.0.1", "172.16.0.5"],
    "dst_ip": ["8.8.8.8", "8.8.4.4", "1.1.1.1", "8.8.8.8", "10.0.0.3"],
    "protocol": ["TCP", "UDP", "TCP", "TCP", "ICMP"],
    "bytes": [120, 340, 560, 80, 10],
    "threat": ["benign", "benign", "malicious", "benign", "malicious"]
}
df = pd.DataFrame(data)
print("🔹 DataFrame:\n", df)
```

```python
print(df.head(3))     # first rows
print(df.tail(2))     # last rows
print(df.info())      # datatypes, nulls
print(df.describe())  # numeric stats
print(df["protocol"].value_counts())  # freq count
```

```python
print(df[["src_ip", "protocol"]])   # select cols
print(df.loc[0])                    # select row by index
print(df.iloc[1:3])                 # select by row range
print(df[df["protocol"] == "TCP"])  # filter condition
print(df[df["bytes"] > 100])        # filter numeric
```

```python
print(df.sort_values("bytes", ascending=False))
```

```python
print(df.groupby("protocol")["bytes"].sum())
print(df.groupby("threat")["bytes"].mean())
```

```python
df2 = df.copy()
df2.loc[2, "protocol"] = None  # add missing
print(df2.isnull().sum())      # count nulls
print(df2.fillna("UNKNOWN"))   # fill missing
```

```python
blacklist = pd.DataFrame({
    "ip": ["192.168.1.10", "172.16.0.5"],
    "category": ["C2 server", "Scanner"]
})
merged = df.merge(blacklist, left_on="src_ip", right_on="ip", how="left")
print(merged)
```
