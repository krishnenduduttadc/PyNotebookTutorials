```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Sample: load a small NSL-KDD dataset (replace with your own path)
url = "https://raw.githubusercontent.com/defcom17/NSL_KDD/master/KDDTrain+.txt"
cols = [f"f{i}" for i in range(41)] + ["label"]
df = pd.read_csv(url, names=cols)

# Encode categorical features
for col in df.select_dtypes(include=['object']):
    df[col] = LabelEncoder().fit_transform(df[col])

X = df.drop("label", axis=1)
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

clf = RandomForestClassifier(n_estimators=100, n_jobs=-1)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
import numpy as np
from sklearn.svm import SVC

# Fake malware dataset: 1000 samples, 20 features
X = np.random.rand(1000, 20)
y = np.random.choice([0,1], size=1000)  # 0 = benign, 1 = malware

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

clf = SVC(kernel='rbf')
clf.fit(X_train, y_train)
print("Malware detection accuracy:", clf.score(X_test, y_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

emails = [
    "Update your bank account immediately or it will be closed",
    "Meeting scheduled tomorrow at 10 AM",
    "Your PayPal account is compromised, click here to reset",
    "Lunch at 1 PM in cafeteria"
]
labels = [1, 0, 1, 0]  # 1 = phishing, 0 = safe

vec = TfidfVectorizer()
X = vec.fit_transform(emails)

clf = LogisticRegression()
clf.fit(X, labels)
print(clf.predict(vec.transform(["Urgent: reset your password now"])))
```

```python
from sklearn.ensemble import IsolationForest

# Example: system call counts, network flow stats
X = np.random.rand(1000, 10)
clf = IsolationForest(contamination=0.05)
clf.fit(X)
y_pred = clf.predict(X)  # -1 = anomaly, 1 = normal

print("Anomalies detected:", (y_pred == -1).sum())
```

```python
import numpy as np
from sklearn.ensemble import IsolationForest

# Example: system call counts, network flow stats (10 features)
X = np.random.rand(1000, 10)

# Train Isolation Forest
iso_clf = IsolationForest(contamination=0.05, random_state=42)
iso_clf.fit(X)

# Predict anomalies (-1 = anomaly, 1 = normal)
y_pred = iso_clf.predict(X)

print("Total anomalies detected:", (y_pred == -1).sum())
print("Normal samples:", (y_pred == 1).sum())
```
