1. Hashing & Integrity

Used to verify files, passwords, or logs.

```python
import hashlib

data = b"Important log entry"

# MD5 (fast but weak, collisions exist)
print("MD5:", hashlib.md5(data).hexdigest())

# SHA-256 (secure, widely used)
print("SHA-256:", hashlib.sha256(data).hexdigest())
```

2. Symmetric Encryption (AES)

Used for protecting sensitive logs or configs.

```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes

key = get_random_bytes(16)   # 128-bit key
cipher = AES.new(key, AES.MODE_EAX)

plaintext = b"Secret attack plan"
ciphertext, tag = cipher.encrypt_and_digest(plaintext)

print("Ciphertext:", ciphertext)

# Decrypt
cipher_dec = AES.new(key, AES.MODE_EAX, nonce=cipher.nonce)
decrypted = cipher_dec.decrypt(ciphertext)
print("Decrypted:", decrypted)
```

3. Asymmetric Encryption (RSA)

Used in TLS/SSL, digital signatures, key exchange.

```python
from Crypto.PublicKey import RSA
from Crypto.Cipher import PKCS1_OAEP

# Generate key pair
key = RSA.generate(2048)
private_key = key
public_key = key.publickey()

# Encrypt with public key
cipher_rsa = PKCS1_OAEP.new(public_key)
ciphertext = cipher_rsa.encrypt(b"Malware sample hash")

# Decrypt with private key
cipher_rsa_dec = PKCS1_OAEP.new(private_key)
print("Decrypted:", cipher_rsa_dec.decrypt(ciphertext))
```

4. Password Hashing (with Salt)

Storing raw passwords is insecure.

```python
import bcrypt

password = b"mypassword123"
hashed = bcrypt.hashpw(password, bcrypt.gensalt())

print("Hashed password:", hashed)
print("Check password:", bcrypt.checkpw(password, hashed))
```

5. Port Scanning (Recon)

Like a mini-nmap.

```python
import socket

target = "scanme.nmap.org"
ports = [21, 22, 80, 443]

for port in ports:
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(1)
    result = sock.connect_ex((target, port))
    if result == 0:
        print(f"Port {port} is OPEN")
    else:
        print(f"Port {port} is CLOSED")
    sock.close()
```

7. Basic Anomaly Detection

Catch unusual behavior with statistics.

```python
import numpy as np

# Bytes transferred per session
traffic = np.array([100, 120, 110, 115, 6000])  # last one is odd

mean, std = traffic.mean(), traffic.std()
threshold = mean + 3*std

anomalies = traffic[traffic > threshold]
print("Anomalous traffic:", anomalies)
```

8. Digital Signatures (Integrity + Non-Repudiation)

Used to prove authenticity of messages/files.

```python
from Crypto.PublicKey import RSA
from Crypto.Signature import pkcs1_15
from Crypto.Hash import SHA256

# Generate keys
key = RSA.generate(2048)
private_key = key
public_key = key.publickey()

message = b"Critical system update"
h = SHA256.new(message)

# Sign with private key
signature = pkcs1_15.new(private_key).sign(h)

# Verify with public key
try:
    pkcs1_15.new(public_key).verify(h, signature)
    print("✅ Signature valid")
except (ValueError, TypeError):
    print("❌ Invalid signature")
```

12. Password Cracking (Dictionary Attack)

```python
import hashlib

def crack_sha256(hash_value, wordlist):
    for word in wordlist:
        if hashlib.sha256(word.encode()).hexdigest() == hash_value:
            return word
    return None

hash_val = hashlib.sha256(b"admin123").hexdigest()
dictionary = ["123456", "password", "admin123", "qwerty"]

result = crack_sha256(hash_val, dictionary)
print("Password found:", result)
```

14. File Integrity Monitoring

```python
import os, hashlib

def file_hash(path):
    with open(path, "rb") as f:
        return hashlib.sha256(f.read()).hexdigest()

baseline = file_hash("important.txt")
print("Baseline hash:", baseline)

# Later...
if baseline != file_hash("important.txt"):
    print("🚨 ALERT: File modified!")
else:
    print("✅ File unchanged")
```

15. Honeypot Simulation (Fake Service Trap)

```python
import socket

host = "0.0.0.0"
port = 2222  # fake SSH port

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind((host, port))
server.listen(5)
print(f"Honeypot running on port {port}...")

while True:
    client, addr = server.accept()
    print(f"🚨 Connection attempt from {addr}")
    client.send(b"SSH-2.0-OpenSSH_8.2\r\n")
    client.close()
```

18. JWT (JSON Web Token) Verification & Forgery Prevention

JWTs are used in APIs for authentication. Attackers sometimes forge weakly signed tokens.

```python
import jwt
from jwt.exceptions import InvalidSignatureError, ExpiredSignatureError
import datetime

secret = "super_secret_key"

# Legit token creation
payload = {"user": "admin", "exp": datetime.datetime.utcnow() + datetime.timedelta(seconds=10)}
token = jwt.encode(payload, secret, algorithm="HS256")

print("Generated JWT:", token)

# Verifying JWT
try:
    decoded = jwt.decode(token, secret, algorithms=["HS256"])
    print("Valid JWT:", decoded)
except ExpiredSignatureError:
    print("Token expired!")
except InvalidSignatureError:
    print("Invalid signature!")
```

19. Port Knocking Simulation

Port knocking hides services behind a sequence of connection attempts.

```python
import socket

secret_sequence = [1234, 2345, 3456]
knock_log = []

def knock(port):
    knock_log.append(port)
    if knock_log[-len(secret_sequence):] == secret_sequence:
        print("🔓 Access granted!")
    else:
        print(f"Knock on port {port}")

# Simulate knocking
knock(1234)
knock(2345)
knock(3456)
```

20. Honeytokens (Fake Credentials Trap)

Honeytokens detect intruders when fake credentials are used.

```python
fake_db = {"admin": "Password123!"}  # honeytoken

def login(user, password):
    if user in fake_db and fake_db[user] == password:
        print("🚨 ALERT! Honeytoken triggered.")
    else:
        print("Login failed.")

login("admin", "Password123!")  # Attack attempt
```

21. Ransomware-Style File Encryption

⚠️ Educational use only — simulating how ransomware encrypts files.

```python
from cryptography.fernet import Fernet
import os

key = Fernet.generate_key()
cipher = Fernet(key)

# Encrypt all .txt files in folder
for fname in os.listdir("."):
    if fname.endswith(".txt"):
        with open(fname, "rb") as f:
            data = f.read()
        encrypted = cipher.encrypt(data)
        with open(fname + ".enc", "wb") as f:
            f.write(encrypted)
        print(f"Encrypted {fname}")
```

22. Brute Force Detection (Login Attempts)

Detecting repeated failed login attempts.

```python
from collections import defaultdict

attempts = defaultdict(int)

def login(user, password):
    attempts[user] += 1
    if attempts[user] > 3:
        print(f"🚨 User {user} locked out after {attempts[user]} attempts!")
    else:
        print(f"Login failed for {user}")

for _ in range(5):
    login("admin", "wrongpass")
```

25. API Rate Limiting

```python
import time
from collections import deque

requests_log = deque(maxlen=5)  # store last 5 requests
time_window = 10  # seconds

def api_call(user):
    now = time.time()
    requests_log.append(now)
    # remove old requests
    while requests_log and now - requests_log[0] > time_window:
        requests_log.popleft()
    if len(requests_log) > 5:
        print("🚨 Rate limit exceeded!")
    else:
        print("API request processed.")

for _ in range(7):
    api_call("user1")
```

26. ARP Spoofing Detection

Detecting if multiple MACs claim same IP.

```python
arp_table = {"192.168.1.1": "AA:BB:CC:DD:EE:FF"}

def detect_arp(ip, mac):
    if ip in arp_table and arp_table[ip] != mac:
        print(f"🚨 ARP Spoofing detected for {ip}! Legit={arp_table[ip]}, Fake={mac}")
    else:
        arp_table[ip] = mac
        print(f"ARP entry updated: {ip} -> {mac}")

detect_arp("192.168.1.1", "AA:BB:CC:DD:EE:FF")
detect_arp("192.168.1.1", "11:22:33:44:55:66")  # spoof attempt
```
