```python
import re

text = "My number is 12345."
result = re.sub(r"\d", "*", text)
print(result)   # My number is *****.
```

```python
text = "Doe, John"
result = re.sub(r"(\w+), (\w+)", r"\2 \1", text)
print(result)   # John Doe
```

```python
import re

# Define simple pattern-response rules
patterns = [
    (r'I need (.*)',
     "Why do you need {0}?"),
    (r'Why don\'?t you ([^\?]*)\??',
     "Do you really think I don't {0}?"),
    (r'Why can\'?t I ([^\?]*)\??',
     "What makes you think you should be able to {0}?"),
    (r'I can\'?t (.*)',
     "How do you know you can't {0}?"),
    (r'I am (.*)',
     "How long have you been {0}?"),
    (r'I\'?m (.*)',
     "Why are you {0}?"),
    (r'Are you (.*)',
     "Why does it matter whether I am {0}?"),
    (r'What (.*)',
     "Why do you ask?"),
    (r'How (.*)',
     "How do you suppose?"),
    (r'(.*) mother(.*)',
     "Tell me more about your mother."),
    (r'(.*) father(.*)',
     "Tell me more about your father."),
    (r'(.*)\?',
     "Why do you ask that?"),
    (r'(.*)',
     "Please tell me more.")
]

def eliza_chat():
    print("ELIZA: Hello, I am ELIZA. How can I help you today?")
    while True:
        user_input = input("You: ")
        if user_input.lower() in ["quit", "exit", "bye"]:
            print("ELIZA: Goodbye! Take care.")
            break
        for pattern, response in patterns:
            match = re.match(pattern, user_input, re.IGNORECASE)
            if match:
                # Replace {0} with the first group, etc.
                reply = response.format(*[g for g in match.groups()])
                print("ELIZA:", reply)
                break

# Run chatbot
eliza_chat()
```

```python
import re

text = "apple123 banana456 grape789"

# Match word only if followed by digits
pattern = r"\w+(?=\d+)"
result = re.findall(pattern, text)
print(result)  # ['apple', 'banana', 'grape']
```

```python
#negative Lookahead assertions
```

```python
text = "foo1 foo foo2 bar bar3"

# Match 'foo' only if NOT followed by digits
pattern = r"foo(?!\d)"
result = re.findall(pattern, text)
print(result)  # ['foo']
```
