```python
# Dependency parsing with spaCy (small model)
# Run once if needed:
# !pip install spacy
# !python -m spacy download en_core_web_sm

import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")   # or "en_core_web_trf" for transformer-backed parser

text = "The quick brown fox jumped over the lazy dog while the farmer watched the scene closely."

doc = nlp(text)

# 1) Print token, head, and dependency label (one token per line)
print("Token\tHead\tDep\tLemma")
for token in doc:
    print(f"{token.text}\t{token.head.text}\t{token.dep_}\t{token.lemma_}")

# 2) Extract common relations: subjects, objects, prepositions with objects
subjects = [tok for tok in doc if tok.dep_ in ("nsubj", "nsubjpass")]
objects  = [tok for tok in doc if tok.dep_ in ("dobj", "obj", "pobj")]
print("\nSubjects:", [t.text for t in subjects])
print("Objects:", [t.text for t in objects])

# 3) Get subtree (e.g., full noun phrase) for a given token (example: subject)
for subj in subjects:
    subtree_text = " ".join([t.text for t in subj.subtree])
    print(f"Full subtree for subject '{subj.text}': {subtree_text}")

# 4) Visualize dependency parse inline (Jupyter)
# Use displacy.render(doc, style="dep", jupyter=True, options={"distance":110})
display = False  # set to True to show visualization in notebook
if display:
    displacy.render(doc, style="dep", jupyter=True, options={"distance":110})
```
