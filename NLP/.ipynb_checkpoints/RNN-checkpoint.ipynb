```python
import torch
import torch.nn as nn
```

```python
# Sequence of numbers
seq = torch.tensor([[1.0], [2.0], [3.0], [4.0]])  # shape: (seq_len, input_size)
seq = seq.view(len(seq), 1, -1)  # (seq_len, batch, input_size)
```

```python
input_size = 1
hidden_size = 5
rnn = nn.RNN(input_size, hidden_size, batch_first=False)
```

```python
h0 = torch.zeros(1, 1, hidden_size)  # initial hidden state (num_layers, batch, hidden)
out, hn = rnn(seq, h0)

print("Output shape:", out.shape)
print("Hidden state shape:", hn.shape)
```

```python
lstm = nn.LSTM(input_size, hidden_size)
out, (hn, cn) = lstm(seq)

print("LSTM output shape:", out.shape)
print("LSTM hidden:", hn.shape, "Cell state:", cn.shape)
```

```python
# Toy dataset: y = x + 1
X = torch.tensor([[1],[2],[3],[4]], dtype=torch.float).view(4,1,1)
y = torch.tensor([[2],[3],[4],[5]], dtype=torch.float).view(4,1,1)

rnn = nn.RNN(1, 10)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(rnn.parameters(), lr=0.01)

h0 = torch.zeros(1, 1, 10)

for epoch in range(300):
    optimizer.zero_grad()
    out, h0 = rnn(X, h0.detach())
    loss = criterion(out, y)
    loss.backward()
    optimizer.step()

out
```
