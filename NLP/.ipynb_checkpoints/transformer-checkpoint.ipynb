```python
import torch
from transformers import AutoTokenizer, AutoModel
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
model = AutoModel.from_pretrained("distilbert-base-uncased")
```

```python
text = "Transformers are powerful models for NLP."
inputs = tokenizer(text, return_tensors="pt")
inputs
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
outputs.last_hidden_state.shape
```

```python
emb = outputs.last_hidden_state[0]
emb
```

```python
from transformers import BertTokenizer, BertModel

tok = BertTokenizer.from_pretrained("bert-base-uncased")
ber = BertModel.from_pretrained("bert-base-uncased", output_attentions=True)

inp = tok("Transformers learn dependencies", return_tensors="pt")

with torch.no_grad():
    out = ber(**inp)

attentions = out.attentions
len(attentions), attentions[0].shape
```

```python
import torch
import torch.nn.functional as F

x = torch.randn(1, 5, 4)  # (batch, seq, features)

Q = x @ torch.randn(4, 4)
K = x @ torch.randn(4, 4)
V = x @ torch.randn(4, 4)

scores = Q @ K.transpose(-2, -1) / (4 ** 0.5)
weights = F.softmax(scores, dim=-1)
output = weights @ V

output
```
