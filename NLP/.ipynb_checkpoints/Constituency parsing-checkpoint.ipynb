```python
# Constituency parsing with spaCy + benepar
# Run these install commands *once* in your notebook/kernel if needed:
# !pip install spacy benepar nltk
# !python -m spacy download en_core_web_sm

import spacy
import benepar   # neural constituency parser (Berkeley)
from nltk import Tree   # for pretty printing the parse tree

# Load spaCy and attach benepar (download model first time if necessary)
# benepar.download('benepar_en3')   # uncomment to download the benepar model (one-time)
nlp = spacy.load("en_core_web_sm")
if "benepar" not in nlp.pipe_names:
    nlp.add_pipe("benepar", config={"model": "benepar_en3"})

# Example sentence
text = "The quick brown fox jumped over the lazy dog while the farmer watched."

doc = nlp(text)

# Each sentence now has a constituency parse available via the ._.parse_string accessor
for i, sent in enumerate(doc.sents, 1):
    parse_str = sent._.parse_string          # bracketed parse like "(S (NP ...) (VP ...))"
    print(f"\nSentence {i} parse (bracketed):\n{parse_str}\n")
    # Convert to nltk.Tree for pretty printing / tree operations
    tree = Tree.fromstring(parse_str)
    tree.pretty_print()                      # prints a readable tree in the notebook/console

    # Examples of useful extractions:
    # 1) All noun phrase (NP) subtrees
    nps = [" ".join(t.leaves()) for t in tree.subtrees(lambda t: t.label() == "NP")]
    print("Noun Phrases (NP):", nps)

    # 2) Top-level constituents
    top_labels = [child.label() for child in tree]
    print("Top-level constituent labels:", top_labels)
```
