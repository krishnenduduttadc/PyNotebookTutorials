# Edit Distance (Levenshtein Distance)

The **edit distance** between two strings $s_1$ and $s_2$ is the minimum number of operations required to transform $s_1$ into $s_2$.  
Allowed operations are:
- **Insertion**
- **Deletion**
- **Substitution**

---

## Dynamic Programming Formulation

Let $dp[i][j]$ represent the minimum edit distance between the first $i$ characters of $s_1$ and the first $j$ characters of $s_2$.

### Base Cases:
\[
dp[i][0] = i \quad \text{(convert $s_1[0..i-1]$ to empty string using $i$ deletions)}
\]
\[
dp[0][j] = j \quad \text{(convert empty string to $s_2[0..j-1]$ using $j$ insertions)}
\]

### Transition:
For $i > 0, j > 0$:
\[
dp[i][j] =
\begin{cases}
dp[i-1][j-1], & \text{if } s_1[i-1] = s_2[j-1] \\
1 + \min
\begin{cases}
dp[i-1][j-1] & \text{(substitution)} \\
dp[i-1][j]   & \text{(deletion)} \\
dp[i][j-1]   & \text{(insertion)}
\end{cases}, & \text{otherwise}
\end{cases}
\]

### Final Answer:
\[
\text{EditDistance}(s_1, s_2) = dp[|s_1|][|s_2|]
\]


```python
def edit_distance_visual(s1, s2):
    m, n = len(s1), len(s2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    # Initialize base cases
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    # Helper to print DP table
    def print_table():
        print("    ", "  ".join([" "] + list(s2)))
        for i in range(m + 1):
            row_label = " " if i == 0 else s1[i-1]
            print(row_label, dp[i])
        print("-" * 40)

    print("Initial DP table:")
    print_table()

    # Fill DP table
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]  # No operation
            else:
                replace = 1 + dp[i - 1][j - 1]
                delete = 1 + dp[i - 1][j]
                insert = 1 + dp[i][j - 1]
                dp[i][j] = min(replace, delete, insert)

            print(f"After processing s1[{i-1}]='{s1[i-1]}' and s2[{j-1}]='{s2[j-1]}':")
            print_table()

    print("Final Edit Distance:", dp[m][n])
    return dp[m][n]


# Example usage
edit_distance_visual("cat", "cut")
```
