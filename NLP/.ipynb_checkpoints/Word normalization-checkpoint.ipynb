1. Lowercasing

```python
text = "Running FAST in the U.S.A."
print(text.lower())
# running fast in the u.s.a.
```

2. Removing Punctuation / Special Characters

```python
import re
text = "Hello!!! How's it going??"
cleaned = re.sub(r"[^a-zA-Z0-9\s]", "", text)
print(cleaned)
# Hello Hows it going
```

3. Tokenization + Stopword Removal (using NLTK)

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "The quick brown fox was running faster than the dogs"
tokens = word_tokenize(text)

# remove stopwords
filtered = [w for w in tokens if w.lower() not in stopwords.words("english")]
print(filtered)
# ['quick', 'brown', 'fox', 'running', 'faster', 'dogs']
```

4. Stemming (rule-based reduction)

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
words = ["running", "runner", "ran", "runs"]
print([ps.stem(w) for w in words])
# ['run', 'runner', 'ran', 'run']
```

5. Lemmatization (dictionary-based reduction)

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
words = ["running", "better", "cars"]
print([lemmatizer.lemmatize(w, pos="v") for w in words])  # verb form
# ['run', 'good', 'car']
```

6. Unicode Normalization (handling accented chars)

```python
import unicodedata

text = "café"
normalized = unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("utf-8")
print(normalized)
# cafe
```

🔹 When to Use What?

Stemming → Fast, rule-based, but can give weird results.
Lemmatization → Slower but accurate (uses dictionary + POS tags).
Unicode Normalization → Important for multilingual text.
Stopword Removal + Lowercasing → Always a good preprocessing step.
