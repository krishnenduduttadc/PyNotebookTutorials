```python
# Minimal Dialogue Manager Demo (intent classification + rule-based policy)
# Install once:
# !pip install scikit-learn

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# ===== 1. TRAIN A TINY INTENT CLASSIFIER =====

texts = [
    "hi", "hello", "hey",
    "bye", "goodbye", "see you",
    "book a flight", "I want to fly", "flight booking",
    "what is the weather", "weather today", "tell me weather"
]
labels = [
    "greet", "greet", "greet",
    "bye", "bye", "bye",
    "book_flight", "book_flight", "book_flight",
    "weather", "weather", "weather"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

clf = LogisticRegression()
clf.fit(X, labels)

# ===== 2. DIALOGUE STATE TRACKER =====

dialogue_state = {
    "last_intent": None,
    "slots": {}
}

# ===== 3. POLICY: CHOOSE NEXT ACTION =====

def policy(intent, state):
    if intent == "greet":
        return "Hello! How can I help you today?"

    if intent == "bye":
        return "Goodbye! Have a great day."

    if intent == "weather":
        return "Please tell me your city."

    if intent == "book_flight":
        return "Sure, I can book a flight. Where would you like to go?"

    return "I'm not sure I understood that."

# ===== 4. Dialogue Loop =====

def dialogue_turn(user_input):
    # Predict intent
    X_test = vectorizer.transform([user_input])
    intent = clf.predict(X_test)[0]

    # Update dialogue state
    dialogue_state["last_intent"] = intent

    # Decide system action
    response = policy(intent, dialogue_state)

    return intent, response


# ===== 5. RUN SOME SAMPLE TURNS =====
print(dialogue_turn("hello"))
print(dialogue_turn("weather today"))
print(dialogue_turn("book a flight"))
print(dialogue_turn("bye"))
```
