```python
import torch
import torch.nn as nn

class MiniTransformer(nn.Module):
    def __init__(self, depth=2, width=128):
        super().__init__()
        self.layers = nn.ModuleList([
            nn.TransformerEncoderLayer(
                d_model=width, nhead=4, dim_feedforward=4*width
            ) for _ in range(depth)
        ])
        
    def forward(self, x):
        for layer in self.layers:
            x = layer(x)
        return x

def count_params(model):
    return sum(p.numel() for p in model.parameters())

# Small model
small = MiniTransformer(depth=2, width=128)
print("Small model params:", count_params(small))

# Scaled-up model
big = MiniTransformer(depth=8, width=512)
print("Big model params:", count_params(big))
```
