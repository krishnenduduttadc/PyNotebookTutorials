NLP style lexical analysis

```python
# Lexical Analysis (Tokenization) using spaCy
# Install once:
# !pip install spacy
# !python -m spacy download en_core_web_sm

import spacy
nlp = spacy.load("en_core_web_sm")

text = "The price of Bitcoin increased by 5.4% on 7th December, 2025!"

doc = nlp(text)

print("Tokens:")
for token in doc:
    print(f"{token.text:12} →  Type: {token.pos_}")
```

Compiler-style Lexical Analysis (Regex-based Tokenizer)

```python
import re

# Token specification: (token_name, regex)
token_spec = [
    ("NUMBER",   r'\d+(\.\d+)?'),
    ("ID",       r'[A-Za-z_]\w*'),
    ("ASSIGN",   r'='),
    ("OP",       r'[+\-*/]'),
    ("LPAREN",   r'\('),
    ("RPAREN",   r'\)'),
    ("SKIP",     r'[ \t]+'),
    ("MISMATCH", r'.'),
]

token_regex = "|".join(f"(?P<{name}>{pattern})" for name, pattern in token_spec)

def lex(code):
    tokens = []
    for m in re.finditer(token_regex, code):
        kind = m.lastgroup
        value = m.group()
        if kind == "SKIP":
            continue
        elif kind == "MISMATCH":
            raise RuntimeError(f"Unexpected character: {value}")
        tokens.append((kind, value))
    return tokens

# Example code input
code = "price = 20 + taxRate * 2.5"

print("Tokens:")
print(lex(code))
```
