```python
import nltk
from nltk.util import ngrams
from collections import Counter, defaultdict

nltk.download("punkt")
```

```python
text = "I love machine learning. I love deep learning. I love coding."

tokens = nltk.word_tokenize(text.lower())
tokens
```

```python
unigrams = list(ngrams(tokens, 1))
unigrams[:5]
```

```python
bigrams = list(ngrams(tokens, 2))
bigrams[:5]
```

```python
trigrams = list(ngrams(tokens, 3))
trigrams[:5]
```

```python
uni_counts = Counter(unigrams)
total_uni = sum(uni_counts.values())

def unigram_prob(word):
    return uni_counts[(word,)] / total_uni
```

```python
bi_counts = Counter(bigrams)
uni_counts2 = Counter(tokens)

def bigram_prob(w1, w2):
    return bi_counts[(w1, w2)] / uni_counts2[w1]
```

```python
tri_counts = Counter(trigrams)
bi_counts2 = Counter(list(ngrams(tokens, 2)))

def trigram_prob(w1, w2, w3):
    return tri_counts[(w1, w2, w3)] / bi_counts2[(w1, w2)]
```

```python
unigram_prob("love")
```

```python
bigram_prob("love", "machine")
```

```python
trigram_prob("i", "love", "machine")
```

```python
import random

def generate_bigram(start_word, length=10):
    word = start_word
    result = [word]
    
    for _ in range(length):
        # collect next words that can follow this word
        candidates = [(w2, c) for (w1, w2), c in bi_counts.items() if w1 == word]
        if not candidates:
            break
        words, counts = zip(*candidates)
        word = random.choices(words, weights=counts)[0]
        result.append(word)
    return " ".join(result)

generate_bigram("i", 8)
```
