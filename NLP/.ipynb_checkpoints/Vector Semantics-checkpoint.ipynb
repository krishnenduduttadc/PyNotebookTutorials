```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["I love NLP", "NLP loves machine learning", "I enjoy learning"]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)

vectorizer.get_feature_names_out(), X.toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X2 = tfidf.fit_transform(docs)

tfidf.get_feature_names_out(), X2.toarray()
```

```python
from gensim.models import Word2Vec

sentences = [
    ["I", "love", "NLP"],
    ["NLP", "loves", "machine", "learning"],
    ["I", "enjoy", "learning"]
]

model = Word2Vec(sentences, vector_size=50, window=3, min_count=1)
```

```python
model.wv["learning"]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

v1 = model.wv["love"]
v2 = model.wv["loves"]

cosine_similarity([v1], [v2])
```

```python
model.wv.most_similar("learning")
```
