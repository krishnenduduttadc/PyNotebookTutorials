Morphology

Morphology is the study of the structure of words and how they are formed from smaller units of meaning.
Example word:
unhappiness
un (prefix, meaning "not")
happy (root word)
ness (suffix, makes it a noun)

🔹 Morphemes

Smallest meaningful unit in a language.
Cannot be divided further without losing meaning.
Types:
Free morphemes → can stand alone as words
e.g., book, run, happy
Bound morphemes → must attach to other morphemes

Prefixes: un-, re-, pre-
Suffixes: -ing, -ed, -ness
Example:
cats = cat (free morpheme) + -s (bound morpheme, plural)

🔹 Lexeme

The base form or set of word forms that represent one meaning.
Think of it as the dictionary entry of a word.
Examples:
Lexeme: RUN
Word forms: run, runs, ran, running
Lexeme: HAPPY
Word forms: happy, happier, happiest

🔹 Morphological Analysis

It’s the process of analyzing a word into its morphemes and identifying the lexeme.
Example: "unhappiness"
Morphemes: un- + happy + -ness
Lexeme: HAPPY
Meaning: "state of not being happy"

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("running", pos="v"))  # run
print(lemmatizer.lemmatize("better", pos="a"))   # good
```

```python
import spacy
#!python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")
doc = nlp("The cats are running happily")

for token in doc:
    print(token.text, "→ lemma:", token.lemma_, "→ pos:", token.pos_, "→ morph:", token.morph)

```
