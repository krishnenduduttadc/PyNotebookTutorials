```python
#Match
```

```python
import re

text = "hello world"
result = re.match(r"hello", text)
print(result.group())   # Output: hello
```

```python
#Search
```

```python
text = "say hello world"
result = re.search(r"hello", text)
print(result.group())   # Output: hello
```

```python
#Find all 
```

```python
text = "cat bat rat mat"
print(r"\bat\b")  # Prints \bat\b (raw string, no backspace)
result = re.findall(r"\bat\b", text)
```

```python
#Email
```

```python
email = "user123@gmail.com"
pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"
print(bool(re.match(pattern, email)))   # True
```

```python
#Find number
```

```python
text = "Order 123, Bill 456, Code 789"
numbers = re.findall(r"\d+", text)
print(numbers)   # ['123', '456', '789']
```

```python
#Split by
```

```python
text = "apple, banana; orange|grape"
result = re.split(r"[;|,]", text)
print(result)   # ['apple', ' banana', ' orange', 'grape']
```
