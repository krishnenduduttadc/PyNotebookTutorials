2. Neural Networks with nn.Module

✔ Linear, ReLU
✔ forward pass

```python
import torch.nn as nn

class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(10, 1)

    def forward(self, x):
        return self.fc(x)
```

3. Training Loop (VERY IMPORTANT)

✔ loss
✔ optimizer
✔ backward + step

```python
model = SimpleNet()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
loss_fn = nn.MSELoss()

for epoch in range(100):
    x = torch.randn(32, 10)
    y = torch.randn(32, 1)

    pred = model(x)
    loss = loss_fn(pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

```python
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)

z = x*y + y**2
z.backward()

print("dz/dx:", x.grad)
print("dz/dy:", y.grad)
```

Linear Regression from scratch

```python
import torch
torch.manual_seed(42)

x = torch.randn(100, 1)
y = 3 * x + 2 + 0.1 * torch.randn(100, 1)
x[:5], y[:5]
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

w, b
y_pred = w * x + b
y_pred[:5]
loss = ((y_pred - y) ** 2).mean()
loss
loss.backward()

print("dw:", w.grad)
print("db:", b.grad)
lr = 0.1

with torch.no_grad():
    w -= lr * w.grad
    b -= lr * b.grad

w.grad = None
b.grad = None


w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

lr = 0.1
epochs = 100

for epoch in range(epochs):
    y_pred = w * x + b
    loss = ((y_pred - y) ** 2).mean()
    
    loss.backward()
    
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad
        
    w.grad = None
    b.grad = None
    
    if epoch % 10 == 0:
        print(f"Epoch {epoch}: loss={loss.item():.4f}")
print("Learned w:", w.item())
print("Learned b:", b.item())
```
