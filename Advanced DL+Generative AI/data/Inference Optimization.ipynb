```python
import time
import statistics

from langchain_community.chat_models import ChatOllama
```

Baseline model (no optimization)

```python
llm = ChatOllama(
    model="llama3",
    temperature=0
)
```

Simple latency benchmark

```python
def benchmark(prompt: str, runs: int = 5):
    times = []

    for _ in range(runs):
        start = time.time()
        llm.invoke(prompt)
        times.append(time.time() - start)

    return {
        "avg": statistics.mean(times),
        "min": min(times),
        "max": max(times)
    }
```

Reduce output tokens (BIGGEST WIN)

```python
short_llm = ChatOllama(
    model="llama3",
    temperature=0,
    num_predict=50  # max tokens
)
```

Temperature = speed (underrated)

```python
fast_llm = ChatOllama(
    model="llama3",
    temperature=0,
    top_k=20,
    top_p=0.9
)
```

Prompt compression (free speed)

```python
long_prompt = """
You are a helpful assistant.
Please carefully and thoroughly explain inference optimization
in great detail, including many examples and explanations.
"""

short_prompt = "Explain inference optimization briefly."

benchmark(long_prompt)
benchmark(short_prompt)
```

Model size comparison (critical)

```python
llm_8b = ChatOllama(model="llama3:8b")
llm_70b = ChatOllama(model="llama3:70b")

def bench_model(model, label):
    start = time.time()
    model.invoke("Summarize inference optimization.")
    print(label, time.time() - start)

bench_model(llm_8b, "8B")
bench_model(llm_70b, "70B")
```
