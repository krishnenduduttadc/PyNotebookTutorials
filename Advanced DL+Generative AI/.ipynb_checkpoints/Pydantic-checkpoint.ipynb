```python
from pydantic import BaseModel, field_validator
import pandas as pd

# Bigger messy dataset (10 rows)
data = [
    {"name": "alice", "age": "30", "email": "ALICE@MAIL.COM "},
    {"name": "bob", "age": "22", "email": " BOB@mail.com"},
    {"name": "charlie", "age": "35", "email": "CHARLIE@mail.COM"},
    {"name": "david", "age": "28", "email": " DAVID@MAIL.com "},
    {"name": "eve", "age": "10", "email": "EVE@MAIL.COM"},
    {"name": "frank", "age": "19", "email": " frank@mail.com "},
    {"name": "grace", "age": "27", "email": "GRACE@mail.com"},
    {"name": "heidi", "age": "31", "email": "HEIDI@MAIL.COM "},
    {"name": "ivan", "age": "45", "email": " IVAN@mail.COM"},
    {"name": "judy", "age": "29", "email": "JUDY@mail.com "}
]

class User(BaseModel):
    name: str
    age: int
    email: str

    @field_validator("name")
    def clean_name(cls, v):
        return v.title()

    @field_validator("email")
    def clean_email(cls, v):
        return v.strip().lower()

# Convert raw → cleaned
cleaned_data = [User(**d).model_dump() for d in data]

# Create DataFrame
df = pd.DataFrame(cleaned_data)

# Show result
df
```

```python
raw_df = pd.DataFrame(data)

print("Raw Data:")
display(raw_df)

print("\nCleaned Data:")
display(df)
```
