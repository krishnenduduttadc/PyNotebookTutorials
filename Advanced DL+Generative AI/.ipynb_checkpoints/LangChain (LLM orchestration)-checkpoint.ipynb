What is LangChain (in one minute)

LangChain is a framework for building applications with LLMs that:

🧩 Chain multiple LLM calls together

🗃️ Plug in tools (search, calculators, APIs)

📚 Add memory (conversation history, summaries)

🔎 Do retrieval-augmented generation (RAG) with documents

🤖 Build agents that decide what tool to use

Think of it as glue code for LLM apps.

Python → LangChain → Ollama → Local model



```python
from langchain_community.llms import Ollama

# Connect to local Ollama
llm = Ollama(
    model="llama3",
    base_url="http://localhost:11434"
)

response = llm.invoke("Explain LangChain in one sentence.")
print(response)
```

prompt

```python
from langchain_community.llms import Ollama
from langchain_core.prompts import PromptTemplate

llm = Ollama(model="llama3")

prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} like I'm 5."
)

formatted_prompt = prompt.format(topic="large language models")

response = llm.invoke(formatted_prompt)
print(response)
```

Chains

```python
from langchain_community.llms import Ollama
from langchain_core.prompts import PromptTemplate
from langchain.chains import LLMChain

llm = Ollama(model="llama3")

prompt = PromptTemplate(
    input_variables=["question"],
    template="Answer clearly: {question}"
)

chain = LLMChain(
    llm=llm,
    prompt=prompt
)

result = chain.invoke({"question": "What is LangChain?"})
print(result["text"])
```

Chat models (conversation-style)

```python
from langchain_community.chat_models import ChatOllama
from langchain_core.messages import HumanMessage

chat = ChatOllama(
    model="llama3",
    base_url="http://localhost:11434"
)

messages = [
    HumanMessage(content="Explain transformers briefly.")
]

response = chat.invoke(messages)
print(response.content)
```

Memory (chat history)

```python
from langchain_community.chat_models import ChatOllama
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory

llm = ChatOllama(model="llama3")

memory = ConversationBufferMemory()

conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=True
)

conversation.predict(input="Hi, my name is Alex.")
conversation.predict(input="What is my name?")
```

Streaming responses (optional but cool)

```python
from langchain_community.llms import Ollama

llm = Ollama(model="llama3")

for chunk in llm.stream("Tell me a short story about a robot."):
    print(chunk, end="", flush=True)

```

Documents → Chunks → Embeddings → Vector DB
                               ↓
User Question → Retrieve chunks → LLM → Answer


```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embedding_model = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)

db = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory="./chroma_db"
)

db.persist()
print("Vector DB created")
```
