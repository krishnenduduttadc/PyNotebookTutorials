An embedding converts text into numbers (vectors) so computers can understand meaning.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

sentences = [
    "I love machine learning",
    "Artificial intelligence is amazing",
    "Dogs are wonderful pets",
    "Cats are cute animals"
]

embeddings = model.encode(sentences)

print(embeddings.shape)
```

Cosine similarity measures how similar two vectors are.

Instead of distance, it measures the angle between vectors.

Value range:

Cosine Similarity	Meaning
1	identical meaning
0.8	very similar
0.5	somewhat related
0	unrelated
-1	opposite

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(embeddings)

print(similarity_matrix)
```

Build a Simple Semantic Search

```python
query = "machine learning and AI"

query_embedding = model.encode([query])

similarities = cosine_similarity(query_embedding, embeddings)

print(similarities)
```

```python
import numpy as np

best_match = np.argmax(similarities)

print("Query:", query)
print("Best match:", sentences[best_match])
```

Visualize Similarity

```python
import pandas as pd

df = pd.DataFrame(similarity_matrix, index=sentences, columns=sentences)
df
```
